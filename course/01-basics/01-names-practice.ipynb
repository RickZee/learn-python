## Core

Return a short label for a value's type. Use `"missing"` when the value is `None`. Otherwise return the type name (`"int"`, `"str"`, `"bool"`, …).

Do it in steps: get the type, read its name, then decide.

<details><summary>Hint: the idea</summary>

`None` is a value with its own type name. Treat that case first so you do not print `NoneType` to the user.
</details>

<details><summary>Hint: the approach</summary>

</details>
