## Core

Return the type name: `int`, `str`, `bool`, or `NoneType`.

<details><summary>Hint: the idea</summary>

Every value has a type. You want that type's name, as text.
</details>

<details><summary>Hint: the approach</summary>

`type(value)` is the type. Ask that object for its `__name__`.
</details>

<details><summary>Hint: near the code</summary>

</details>


```python
def kind(value):
    raise NotImplementedError


check(kind, 3, expect="int")
check(kind, "hello", expect="str")
check(kind, True, expect="bool")
check(kind, None, expect="NoneType")
check(kind, 0, expect="int")
```
