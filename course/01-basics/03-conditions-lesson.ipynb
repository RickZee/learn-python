## The idea

A comparison such as `n < 0` produces a value: `True` or `False`. `if` runs its indented block only when its condition is true. `elif` is checked only when every branch above it was false, and `else` catches whatever is left. `None` means "no value yet". It isn't `False` and it isn't `0`, so ask about it directly with `value is None`.

<details><summary>Explain more</summary>

The colon at the end of `if n < 0:` starts a block. The block is the lines indented under it (4 spaces). `=` stores a value; `==` compares two values. `is` asks whether two names point at the very same object, which is how you test for `None`.
</details>


## See it work

A function that labels a number, and treats a missing value before it compares.


```python
def describe(n: int | None) -> str:
    # 1. Handle the missing value first.
    if n is None:
        return "missing"
    # 2. Check special cases, most specific first.
    if n < 0:
        return "negative"
    # 3. Everything else.
    return "ok"


print(describe(None), describe(-1), describe(0), describe(2))
```

```python
print(3 < 5)
print(0 == False)  # noqa: E712
print(None == False)  # noqa: E711, E712
print(None is None)
print(0 or "default")  # noqa: SIM222
print(0 and "default")  # noqa: SIM223
```

## Why it works this way

`and` and `or` return an operand. `or` returns the first truthy operand, otherwise the last. `and` returns the first falsy, otherwise the last — so `0 or "default"` is `"default"` and `0 and "default"` is `0`. The pattern `name or "fallback"` bites when a real value counts as false: `0` or `""`.

What counts as false comes from the object: `(0).__bool__()` is false; `"".__len__()` is `0`. The text `"0"` has length 1, so it is true. `None` is its own object — test with `is None`.

`bool` is a subclass of `int` (see [names](01-names-lesson.ipynb)): `False == 0`. [Objects, names, references](../../internals/01-objects/01-names-lesson.ipynb) is why `is` checks the object and `==` checks the value. Short-circuiting skips later operands once the result is known.


```python
print("".__len__())
print(bool(""))
print((0).__bool__())
print(bool("0"))
```

## When to use it

Use `if n is None` when `0` and `""` are real data. Use `x or default` only when every false value should be replaced. Prefer a comparison when you need an actual `True` or `False`.


## Common mistake

The last line of the error names the problem: Python can't compare `None` with a number. Check for `None` first.


```python
def describe_wrong(n):
    if n < 0:
        return "negative"
    if n is None:
        return "missing"
    return "ok"


describe_wrong(None)
```
