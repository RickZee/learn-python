## The idea

A name points at a value. `count = 3` makes the name `count` point at the number 3. The value has a type: a number, text, `True` or `False`, or `None`. `type(value).__name__` is the type's name, as text.

<details><summary>Explain more</summary>

`=` stores a value under a name; it does not compare. `None` means "no value yet", not `0` or `"none"`. An unset name raises `NameError`.
</details>

<details><summary>If you already write Java, JavaScript, or C#</summary>

`/` is float division and `//` is integer division. Empty values and `None` are false. Integers have no fixed width.
</details>


## See it work

A small function that stores a title and a copy count, then prints both type names.


```python
def announce(title, copies):
    # 1. Store the title under a name.
    label = title
    # 2. Store the count under a different name.
    amount = copies
    print(label, type(label).__name__)
    print(amount, type(amount).__name__)


announce("notebook", 2)
```

```python
print(type(3).__name__)  # noqa: UP003
print(type("3").__name__)  # noqa: UP003
print(type(3.0).__name__)  # noqa: UP003
ready = True
print(type(ready).__name__)
missing = None
print(type(missing).__name__)
```

<details><summary>Check your answers</summary>

- `3` is an `int`. `"3"` is a `str`. `3.0` is a `float`.
- `True` is a `bool` (subclass of `int`), so `True == 1`.
- `None` prints as `NoneType`.
</details>


`bool` is a subclass of `int`. This cell prints the evidence.


```python
print(isinstance(True, int))
print(True + True)
```

## Why it works this way

A name is a label on an object, not a box that holds a copy. `right = left` stores the same pointer again — the next cell prints `True` for `left is right`. `other` is built by joining two strings, so it is equal and not the same object. [Python Tutor](https://pythontutor.com/visualize.html) shows two arrows into one string, then a second for `other`. [Objects, names, references](../../internals/01-objects/01-names-lesson.ipynb) is that pointer write.

`True` is a `bool`, a subclass of `int`, so `type(True).__name__` says `bool` and `isinstance(True, int)` is true. Rebinding writes one pointer, not a deep copy of the object, so two names can share one large string without doubling memory.


```python
left = "notebook"
right = left
head = "note"
tail = "book"
other = head + tail
print(left is right)
print(left == other)
print(left is other)
```

## When to use it

Add a name for another label on an object you already have. Build a new object when a later change must not show through the first name. Use `is` only for `None`.


## Common mistake

The last line of the error names the problem: the name has not been stored yet. Store a value before you use the name.


```python
print(total)  # noqa: F821
total = 1
print("stored", total)
```
