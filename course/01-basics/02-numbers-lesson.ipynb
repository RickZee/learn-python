## The idea

`/` divides and keeps the fraction, so the result is a `float`. `//` divides and drops the fraction, so you get how many full groups fit. `%` is what is left over after those groups.

<details><summary>Explain more</summary>

An `int` is a whole number. A `float` has a fraction, even when that fraction is `.0`. `10 // 3` is `3`, because three groups of 3 fit into 10. `10 % 3` is `1`, the leftover.
</details>


## See it work

Turn a number of minutes into whole hours and the minutes left over.


```python
def hours_and_minutes(total):
    # 1. How many full hours fit.
    hours = total // 60
    # 2. What is left over.
    minutes = total % 60
    return hours, minutes


print(hours_and_minutes(135))
print(hours_and_minutes(60))
```

## Why it works this way

An `int` grows — `2**100` stays an `int` with as many digits as needed, no fixed-width wrap. [Numbers and strings](../../internals/05-numbers/01-numbers-strings-lesson.ipynb) shows that storage, and the binary fraction a `float` keeps.

A `float` is a fixed-size binary fraction. `0.1` is not a sum of powers of one half, so the stored value is the nearest binary value. Adding two approximations need not land on the value stored for `0.3` — hence `0.1 + 0.2 == 0.3` is false.

`//` floors toward negative infinity. The identity is `n == (n // d) * d + (n % d)`, and the remainder is never negative when `d` is positive. `-7 // 2` is `-4` and `-7 % 2` is `1`, because `-4 * 2 + 1` is `-7`.


```python
print(2**100)
print(type(2**100).__name__)
```

## When to use it

Use `int` for counts, indexes, and money in cents. Use `float` when a tiny relative error is fine and you will not test with `==`. Use `//` and `%` together for full groups and a non-negative remainder. Prefer `int(n / d)` only when you truly want to chop toward zero.


## Common mistake

`int(n / d)` chops toward zero. People use it when they wanted `//`. On positives the two agree. On negatives they do not. `-7` chopped toward zero is `-3`. Floor division is `-4`. The silent wrong answer is `-3` when the rest of the program expects the floor identity.


```python
def toward_zero(n, d):
    return int(n / d)


print(toward_zero(-7, 2))
print(-7 // 2)
print((-7 // 2) * 2 + (-7 % 2))
```
