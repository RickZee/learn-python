## Debug

This function has the bug from the lesson's Common mistake. Find it, say why it happens in one sentence, then fix it.

`require_age` should only catch NegativeAgeError, not every Exception.


```python
class NegativeAgeError(Exception):
    pass


def require_age(age: int) -> int:
    try:
        if age < 0:
            raise NegativeAgeError("negative")
        return age
    except Exception:
        return 0  # hides TypeError on bad input


try:
    require_age("x")  # type: ignore[arg-type]
except TypeError:
    pass
else:
    raise AssertionError("TypeError must not be swallowed")
assert require_age(3) == 3
try:
    require_age(-1)
except NegativeAgeError:
    pass
else:
    raise AssertionError("negative age should raise NegativeAgeError")
```
