## Warm-up

These lines of `Point.__init__` are scrambled. Rebuild them with correct indentation.


```python
#         self.y = y
#     def __init__(self, x: int, y: int):
#         self.x = x
# class Point:


class Point:
    def __init__(self, x: int, y: int):
        raise NotImplementedError

    def moved(self, dx: int, dy: int) -> "Point":
        raise NotImplementedError


check(lambda p: p.x, Point(1, 2), expect=1)
check(lambda p: p.y, Point(1, 2), expect=2)
```

## Core

Store `x` and `y`. `moved` returns a new `Point` shifted by `dx` and `dy`.

<details><summary>Hint: the idea</summary>

The new point's coordinates are the old ones plus the shift.
</details>

<details><summary>Hint: the approach</summary>

Do not assign to `self.x`. Call `Point(...)` with the sums.
</details>

<details><summary>Hint: near the code</summary>

`return Point(self.x + dx, self.y + dy)`
</details>


```python
def _shift_x(point):
    return point.moved(3, 4).x


def _shift_y(point):
    return point.moved(3, 4).y


check(_shift_x, Point(1, 2), expect=4)
check(_shift_y, Point(1, 2), expect=6)
assert Point(1, 2).moved(3, 4).x == 4 and Point(1, 2).x == 1, "old point unchanged"
```

## Stretch

Return a copy with the same coordinates.


```python
def _copy_x(point):
    return point.copy().x


check(_copy_x, Point(4, 7), expect=4)
assert Point(4, 7).copy() is not Point(4, 7), "copy is a new object"
```
