## Core

`Step.add` adds one extra through `super()`. `bumped` returns `Step().add(1)`.

<details><summary>Hint: the idea</summary>

The subclass calls the parent with `n + 1`.
</details>

<details><summary>Hint: the approach</summary>

Copy the `Counter` and `Step` classes from the lesson, then return `Step().add(1)`.
</details>

<details><summary>Hint: near the code</summary>

`return super().add(n + 1)`
</details>


```python
class Counter:
    def __init__(self) -> None:
        self.value = 0

    def add(self, n: int) -> int:
        self.value += n
        return self.value


class Step(Counter):
    def add(self, n: int) -> int:
        raise NotImplementedError


def bumped() -> int:
    return Step().add(1)


check(bumped, expect=2)
```
