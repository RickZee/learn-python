## Core

Return `Celsius(degrees).fahrenheit`.

<details><summary>Hint: the idea</summary>

The property does the conversion.
</details>

<details><summary>Hint: the approach</summary>

Build a `Celsius` and read `.fahrenheit`.
</details>

<details><summary>Hint: near the code</summary>

`return self.degrees * 9 / 5 + 32` inside the property.
</details>


```python
class Celsius:
    def __init__(self, degrees: float) -> None:
        self.degrees = degrees

    @property
    def fahrenheit(self) -> float:
        raise NotImplementedError


def as_fahrenheit(degrees: float) -> float:
    return Celsius(degrees).fahrenheit


check(as_fahrenheit, 0, expect=32)
check(as_fahrenheit, 100, expect=212)
```

## Debug

This function has the bug from the lesson's Common mistake. Find it, say why it happens in one sentence, then fix it.

`read_f` should read the property without calling it like a method.


```python
class Celsius:
    def __init__(self, degrees: float) -> None:
        self.degrees = degrees

    @property
    def fahrenheit(self) -> float:
        return self.degrees * 9 / 5 + 32


def read_f(degrees: float) -> float:
    value = Celsius(degrees).fahrenheit
    # BUG: treat property like a method — call it again
    if callable(value):
        return value()
    return value - 1  # planted wrong when property already returned a float


from pylearn import check

check(read_f, 0, expect=32.0)
check(read_f, 100, expect=212.0)
```

## Stretch

`Room.area` is width times depth. Return the floor area for a room.


```python
class Room:
    def __init__(self, width: float, depth: float) -> None:
        self.width = width
        self.depth = depth

    @property
    def area(self) -> float:
        raise NotImplementedError


def floor_area(width: float, depth: float) -> float:
    return Room(width, depth).area


check(floor_area, 3, 4, expect=12)
check(floor_area, 2.5, 2, expect=5.0)
```
