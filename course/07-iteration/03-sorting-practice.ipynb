`by_score_then_name` should sort by score ascending, then by name for ties.


```python
def by_score_then_name(rows: list[tuple[str, int]]) -> list[tuple[str, int]]:
    return sorted(rows, key=lambda row: row[1])


check(by_score_then_name, [("b", 1), ("a", 1), ("c", 0)], expect=[("c", 0), ("a", 1), ("b", 1)])
```
