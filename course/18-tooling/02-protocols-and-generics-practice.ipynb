## Core

Define a runtime-checkable `SizedName` protocol that requires a `name: str` attribute. Return whether `obj` matches it.

<details><summary>Hint: the idea</summary>

Use `@runtime_checkable` and `isinstance`.

</details>

<details><summary>Hint: the approach</summary>

Protocol with `name: str`.

</details>

<details><summary>Hint: near the code</summary>

`return isinstance(obj, ___)`.

</details>


```python
from typing import Any, runtime_checkable


@runtime_checkable
class SizedName(Protocol):
    name: str


def has_name(obj: Any) -> bool:
    raise NotImplementedError


class Person:
    def __init__(self, name: str) -> None:
        self.name = name


check(has_name, Person("ada"), expect=True)
check(has_name, object(), expect=False)
```
