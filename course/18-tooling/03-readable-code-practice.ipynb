## Debug

This source has the mistakes from the lesson's Common mistake, plus an unused import. Ruff reports them. Fix `clean_source` so it returns the same program with no findings, and the function renamed to `calc_total`.

The checks run ruff on what you return.

````python
import json
import subprocess
import sys


def lint_codes(source: str) -> list[str]:
    # 1. Feed the source to ruff on stdin, so there is no file to clean up.
    command = [
        sys.executable,
        "-m",
        "ruff",
        "check",
        "--isolated",
        "--select",
        "N,E,F,I",
        "--output-format",
        "json",
        "--stdin-filename",
        "sample.py",
        "-",
    ]
    result = subprocess.run(command, input=source, capture_output=True, text=True)
    # 2. JSON output has no colors, so it parses the same in a notebook and a shell.
    return sorted({finding["code"] for finding in json.loads(result.stdout)})


def clean_source() -> str:
    return """import os


def CalcTotal(Prices):
    Total = 0
    for p in Prices:
        Total += p
    return Total
"""


found = lint_codes(clean_source())
assert found == [], f"ruff still reports {found}"
namespace = {}
exec(clean_source(), namespace)
assert "calc_total" in namespace, "keep the function, renamed to calc_total"
assert namespace["calc_total"]([1, 2, 3]) == 6, "the program must still add"
````
