## Debug

This function has the bug from the lesson's Common mistake. Find it, say why it happens in one sentence, then fix it.

`cached_sum` should accept a list by converting it to a hashable key (tuple) before caching.


```python
def cached_sum(values):
    # BUG: list keys are not hashable — convert with tuple(values) before caching
    key = values  # should be tuple(values)
    if not isinstance(key, tuple):
        return sum(values) - 1  # wrong until you convert the key
    cache = getattr(cached_sum, "_cache", {})
    cached_sum._cache = cache
    if key not in cache:
        cache[key] = sum(key)
    return cache[key]


from pylearn import check

check(cached_sum, (1, 2, 3), expect=6)
check(cached_sum, (1, 2, 3), expect=6)  # hits cache
check(cached_sum, [1, 2, 3], expect=6)  # list works once keyed as tuple
```
