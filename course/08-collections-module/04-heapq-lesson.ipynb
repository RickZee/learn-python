## The idea

`heapq` treats a list as a binary min-heap: the smallest value lives at index `0`. `heapify` rearranges a list into that shape in linear time. `heappop` removes the minimum and restores the invariant. `nsmallest` and `nlargest` are helpers for the k-best cases. There is no max-heap flag; negate numbers when you need the largest.

<details><summary>Explain more</summary>

After `heapify`, the list is not fully sorted. Only the heap rule holds: each parent is smaller than its children. That is enough to pop minima quickly.
</details>


## See it work

Find the smallest scores with a heap.


```python
import heapq


def three_smallest(nums):
    # 1. Heapify once, then pop minima.
    heap = list(nums)
    heapq.heapify(heap)
    return [heapq.heappop(heap) for _ in range(min(3, len(heap)))]


print(three_smallest([5, 1, 4, 2, 3]))
print(heapq.nsmallest(3, [5, 1, 4, 2, 3]))
```

## Why it works this way

A binary heap is a complete tree stored in a list. Index `0` is the root. Children of index `i` sit at `2i+1` and `2i+2`. The min-heap invariant says each parent is less than or equal to its children, so the global minimum is always at the root. `heappop` swaps the last leaf into the root and sifts down. `heappush` sifts up. Those repairs are logarithmic in the heap size. Deep dive: [How list works](../../internals/03-lists/01-dynamic-array-lesson.ipynb).

`heapify` builds the invariant from the bottom in O(n) time, which is faster than pushing n items one by one into an empty heap. The list is not sorted afterward. Only the parent-child rule holds. Sorting would be O(n log n) and is the wrong tool when you only need repeated minima or a small top-k.

Python's `heapq` is min-only to keep the API small. For a max-heap of numbers, store negations and negate again on the way out. For mixed types or tie-breaks, store tuples so the first field is the priority and later fields break ties. Cost of `nsmallest(k)` is about n log k with a heap of size k, which beats a full sort when k is small relative to n.

The design is "list plus invariant," not a separate heap type. You see the same list object before and after `heapify`. That keeps memory obvious and lets you peek at `heap[0]` without popping. Never treat a slice of the heap list as the sorted prefix. Pop, or call the helpers, when you need the true order of minima.


```python
import heapq

nums = [5, 1, 4, 2, 3]
heap = list(nums)
heapq.heapify(heap)
print(heap)
print(heapq.heappop(heap))
print(heap)
print(heapq.nsmallest(3, nums))
print(heapq.nlargest(2, nums))
```

## When to use it

Use a heap for running minima, top-k, and priority queues. Use `sorted` when you need the full order once. Don't assume `heapify` sorts the list. Don't expect a max-heap mode; negate or use tuples.


## Common mistake

Reading the whole heap as if `heapify` sorted it. Middle values are not in order, so taking `heap[:3]` is not the three smallest.


```python
import heapq

heap = [5, 1, 4, 2, 3]
heapq.heapify(heap)
print(heap[:3])
```

Pop three times, or call `nsmallest`. The slice is not the answer.


```python
import heapq

nums = [5, 1, 4, 2, 3]
print(heapq.nsmallest(3, nums))
```
