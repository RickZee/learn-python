## Debug

This function has the bug from the lesson's Common mistake. Find it, say why it happens in one sentence, then fix it.

`trail` should append `"ok"` only when the division succeeds (else), and always append `"done"` (finally). Failure must not log success.


```python
def trail(n: int) -> list[str]:
    log = []
    try:
        10 // n
    except ZeroDivisionError:
        log.append("bad")
    finally:
        log.append("ok")
        log.append("done")
    return log


check(trail, 2, expect=["ok", "done"])
check(trail, 0, expect=["bad", "done"])
```
