## Debug

This function has the bug from the lesson's Common mistake. Find it, say why it happens in one sentence, then fix it.

`parse_age` should convert text to int and reject negatives. When `int(text)` fails, re-raise with `from` so the original error is chained.


```python
def parse_age(text: str) -> int:
    try:
        n = int(text)
    except ValueError:
        raise ValueError("not an int")
    if n < 0:
        raise ValueError("negative")
    return n


check(parse_age, "3", expect=3)
try:
    parse_age("x")
except ValueError as err:
    assert err.__cause__ is not None or err.__context__ is not None
else:
    raise AssertionError("expected ValueError")
```
