## Debug

This function has the bug from the lesson's Common mistake. Find it, say why it happens in one sentence, then fix it.

`vowels_or_none` should return the vowels found in `text`, or `"none"` if there were none.


````python
def vowels_or_none(text: str) -> str:
    found = ""
    for char in text:
        if char in "aeiou":
            found = found + char
    else:
        found = "none"
    return found


check(vowels_or_none, "cat", expect="a")
check(vowels_or_none, "sky", expect="none")
check(vowels_or_none, "ae", expect="ae")
````
