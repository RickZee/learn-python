## Debug

This function has the bug from the lesson's Common mistake. Find it, say why it happens in one sentence, then fix it.

`label_status` should treat only the text `"ok"` as fine.


````python
def label_status(status: str) -> str:
    match status:
        case ok:  # planted: bare name captures every status
            return "fine"
    return "bad"


check(label_status, "ok", expect="fine")
check(label_status, "missing", expect="bad")
check(label_status, "error", expect="bad")
````
