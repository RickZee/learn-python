## The idea

A docstring is a string as the first statement in a function. Python stores it on `__doc__`. A hint such as `width: int` labels a parameter for readers and for tools. Python does not check hints when the function runs.

<details><summary>Explain more</summary>

Triple quotes are the usual form. `help` and editors read the docstring. Type checkers and libraries such as pydantic can read hints. The call still runs if you pass the wrong type, unless some other tool stops you.
</details>


## See it work

Document a rectangle area helper.


```python
def area(width: int, height: int) -> int:
    """Return width times height."""
    # 1. The docstring is stored on the function.
    # 2. The return value is separate from that string.
    return width * height


print(area(2, 3))
print(area.__doc__)
```

## Why it works this way

The docstring is ordinary program data stored on the function object. `__doc__` reads that attribute. Hints are also stored on the function, in `__annotations__`. The calling convention does not look at them. The body runs with whatever objects you passed. That design keeps the runtime small and leaves checking to tools that opt in. Editors, type checkers, and libraries such as pydantic read the annotations. Tests and `help` read the docstring. Deep dive: [Scopes and closures](../../internals/06-scopes/01-closures-lesson.ipynb).

The cost of a docstring or a hint is a string or a small dict on the function object. They do not run on every call. Skipping runtime checks keeps calls fast and leaves room for custom types.


```python
def area(width: int, height: int) -> int:
    """Return width times height."""
    return width * height


print(area.__annotations__)
print(area(2, 3))
```

## When to use it

Write a one-line docstring for every function another person will call. Add hints when the types are not obvious. Don't rely on hints to validate user input. Check values in the body, or use a library that turns hints into validators.


## Common mistake

Treating a hint as a runtime guard. The call with the wrong type still enters the body. Here the strings concatenate, so nothing raises, and the quiet wrong answer is `"nana"`.


```python
def double(n: int) -> int:
    return n + n


print(double("na"))
```

If you need a real check, write one with ordinary code. The hint can stay for tools. Here a non-int is turned into a clear result you can test, without using `raise` yet.


```python
def double(n: int) -> int:
    if type(n) is not int:
        return 0
    return n + n


print(double(3))
print(double("na"))
```

## Recap

- The docstring is the first string in the function and lives on `__doc__`.
- Hints live on `__annotations__` and are not enforced at run time.
- Tools read both. The runtime runs the body either way.
