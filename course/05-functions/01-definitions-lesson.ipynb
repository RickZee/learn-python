## The idea

A function is a named recipe. `return` hands a value back. Leaving `return` out hands back `None`. Two values come back as one tuple the caller can unpack.

Think of a ticket window. No slip means `None`, not "the work still happened somewhere."

<details><summary>Explain more</summary>

The names in `def tip(bill, rate)` are parameters. The values in `tip(40, 0.2)` are arguments. A function is an object. The name points at that object.
</details>


## See it work

Compute the tip and the total for a bill.


```python
def tip_and_total(bill, rate):
    # 1. Tip is a fraction of the bill.
    tip = bill * rate
    # 2. Hand both numbers back as one tuple.
    return tip, bill + tip


tip, total = tip_and_total(40, 0.2)
print(tip, total)
```

## Check yourself

Answer these before you run the cell.

1. What does `add(2, 3)` return?
2. What do `low` and `high` become after unpacking?
3. What does `silent(2, 3) + 1` do, and why?


```python
def add(a, b):
    return a + b


def bounds(nums):
    return min(nums), max(nums)


def silent(a, b):
    a + b


print(add(2, 3))
low, high = bounds([3, 1, 2])
print(low, high)
print(silent(2, 3) + 1)
```

<details><summary>Answers</summary>

`add` returns `5`. Unpacking stores `1` and `3`. `silent` returns `None`, and `None + 1` raises `TypeError`. The addition inside the function never left the function.
</details>


## Why it works this way

`def` creates a function object and stores a pointer under that name. A call looks up that object and runs its body in a new frame. `return` leaves the frame and gives the caller the value. There is no other channel for a result. Computing a sum and not returning it throws the sum away when the frame ends. That is why a missing `return` is `None`, and why `None + 1` fails on the caller's line. Deep dive: [Scopes and closures](../../internals/06-scopes/01-closures-lesson.ipynb).

Two return values build one tuple. Unpacking needs one name per item. A call costs a new frame plus the work in the body. The function object itself is shared across calls, so printing the name without calling it shows the recipe, not a number.


```python
def add(a, b):
    return a + b


print(type(add).__name__, add.__name__)
print(add(2, 3))
other = add
print(other(4, 1))
```

## When to use it

Use a function when the same steps need a name. Return a tuple for two results. Don't omit `return` and hope the caller reads a local. Locals die with the frame.


## Common mistake

Forgetting `return`. The caller gets `None`, and `None + 1` raises. Read the last line first: `TypeError`. The arrow marks the caller's `+`.


```python
def tip_only(bill, rate):
    bill * rate


print(tip_only(40, 0.2) + 1)
```
