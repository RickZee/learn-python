## Debug

This function has the bug from the lesson's Common mistake. Find it, say why it happens in one sentence, then fix it.

`append_copy` should return a new list with `item` appended, and leave `items` unchanged.


````python
def append_copy(items: list[int], item: int) -> list[int]:
    items.append(item)  # mutates the caller's list
    return items


original = [1, 2]
out = append_copy(original, 3)
assert original == [1, 2], f"caller list changed: {original}"
assert out == [1, 2, 3]
````
