````python
def tagged(n: int, label: str = "n") -> str:
    return label + ":" + str(n)  # planted: label is not keyword-only


assert tagged(3, label="id") == "id:3"
assert tagged.__code__.co_argcount == 1, "label must be keyword-only"
````
