## Debug

This function has the bug from the lesson's Common mistake. Find it, say why it happens in one sentence, then fix it.

`make_handlers` should return one function per label that returns that label. The planted bug closes over the loop variable.


````python
def make_handlers(labels: list[str]):
    handlers = []
    for label in labels:

        def handler():
            return label  # planted: late binding to loop variable

        handlers.append(handler)
    return handlers


hs = make_handlers(["a", "b", "c"])
got = []
for h in hs:
    got.append(h())
assert got == ["a", "b", "c"]
````
