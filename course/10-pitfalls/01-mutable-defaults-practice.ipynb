## Debug

This function has the bug from the lesson's Common mistake. Find it, say why it happens in one sentence, then fix it.

`broken_append` should return a fresh list each call when no argument is passed. The planted bug uses a mutable default.


```python
def broken_append(lst=[]):
    lst.append(1)
    return lst


from pylearn import check

first = broken_append()
second = broken_append()
assert first is not second, "each call needs its own list"
assert first == [1] and second == [1]
```
