`is_zero` should check numeric equality. The planted bug uses `is` for a number.


```python
def is_zero(total: int) -> bool:
    return total is 0  # identity, not equality


from pylearn import check

check(is_zero, 0, expect=True)
check(is_zero, 0 + 1000 - 1000, expect=True)
check(is_zero, 1, expect=False)
```
