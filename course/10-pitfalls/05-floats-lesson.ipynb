## The idea

A `float` is a binary floating-point number. Many decimal fractions have no exact binary representation, so arithmetic that looks exact in base 10 is approximate. `Decimal` stores decimal digits and can add money-like strings exactly. Construct `Decimal` from strings, not from pre-rounded floats.

<details><summary>Explain more</summary>

`Decimal("0.1") + Decimal("0.2")` is exact. `Decimal(0.1)` embeds the already-wrong float. For science and graphics, float is usually the right tool. For currency, prefer `Decimal` or integer cents.
</details>


## See it work

Add money with `Decimal`.


```python
from decimal import Decimal


def money(cents, extra):
    # 1. Build Decimal from strings, not from floats.
    return str(Decimal(cents) + Decimal(extra))


print(0.1 + 0.2)
print(money("0.1", "0.2"))
```

## Why it works this way

Binary floating point represents numbers as a significand times a power of two. Fractions whose denominator is not a power of two often cannot be stored exactly. `0.1` in decimal is a repeating fraction in binary, so the stored value is the closest float. Adding two such approximations compounds the error, which is why `0.1 + 0.2` prints a long remainder and does not equal `0.3`. Deep dive: [Numbers and strings](../../internals/05-numbers/01-numbers-strings-lesson.ipynb).

`Decimal` uses decimal arithmetic with a context for precision. Constructing from a string parses the decimal digits you wrote. Constructing from a float first accepts the binary approximation, then records that already-wrong value. For money, integer cents are another exact approach: store `int` pennies and format on the way out.

Float remains the right default for scientific computing, sensors, and most graphics, where speed and hardware support matter and tiny relative error is expected. Exact decimal identity is the wrong goal there. Choose the representation that matches the question: approximate magnitude, or exact decimal sums.

Evidence: print `0.1 + 0.2`, the failed `== 0.3` check, and the exact `Decimal` sum beside `Decimal(0.1)`. Rounding a float to make a test pass hides the representation error until a later sum drifts. Prefer an exact type at the boundary where the value enters your program. Binary float is a trade of exact decimal digits for speed and range. Know which side of that trade your problem is on before you compare with `==`.


```python
from decimal import Decimal

print(0.1 + 0.2)
print(0.1 + 0.2 == 0.3)
print(Decimal("0.1") + Decimal("0.2"))
print(Decimal(0.1))
print(float(Decimal("0.1") + Decimal("0.2")))
```

## When to use it

Use `Decimal` or integer cents for currency and other exact base-10 totals. Use `float` for science, stats, and performance-sensitive numeric code. Don't build `Decimal` from a float literal when you care about the decimal digits. Parse user-facing decimal text at the edge, then keep one exact representation inside.


## Common mistake

Comparing money with floats and treating a tiny remainder as a real difference, or "fixing" it with another float literal.


```python
total = 0.1 + 0.2
print(total == 0.3)
print(round(total, 1) == 0.3)
```

Add with `Decimal` strings when the sum must be exact in base 10.


```python
from decimal import Decimal

print(Decimal("0.1") + Decimal("0.2") == Decimal("0.3"))
```
