## Debug

This function has the bug from the lesson's Common mistake. Find it, say why it happens in one sentence, then fix it.

`make_handlers` should return callables that each remember their own `n`. The planted bug closes over the loop variable.


```python
def make_handlers(limit: int) -> list:
    handlers = []
    for n in range(limit):
        handlers.append(lambda: n)  # late binding
    return handlers


from pylearn import check

got = [h() for h in make_handlers(3)]
assert got == [0, 1, 2], f"each handler should keep its n; got {got}"
```
