## Debug

This function has the bug from the lesson's Common mistake. Find it, say why it happens in one sentence, then fix it.

`strip_blanks` should return the lines without any empty strings. The planted bug removes items from the list it is looping over.

```python
def strip_blanks(lines: list[str]) -> list[str]:
    for line in lines:
        if line == "":
            lines.remove(line)  # skips the item after each removal
    return lines


assert strip_blanks(["a", "", "", "b"]) == ["a", "b"], "both blanks must go"
check(strip_blanks, ["", "x"], expect=["x"])
```
