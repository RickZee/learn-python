## Debug

This function has the bug from the lesson's Common mistake. Find it, say why it happens in one sentence, then fix it.

`pair_up` should keep every pair when lengths match. The planted bug drops the last pair — the same class of silent length mistake as forgetting `strict=True`.


````python
def pair_up(left: list[int], right: list[str]) -> list[tuple[int, str]]:
    pairs = list(zip(left, right))
    return pairs[:-1] if pairs else pairs  # planted: silent length mistake


from pylearn import check

check(pair_up, [1, 2], ["a", "b"], expect=[(1, "a"), (2, "b")])
check(pair_up, [], [], expect=[])
````
