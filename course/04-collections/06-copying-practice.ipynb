## Debug

This function has the bug from the lesson's Common mistake. Find it, say why it happens in one sentence, then fix it.

`detach_first_row` should return a copy where editing the first row does not touch `grid`.


````python
def detach_first_row(grid: list[list[int]]) -> list[list[int]]:
    copied = list(grid)
    return copied


grid = [[1], [2]]
copied = detach_first_row(grid)
copied[0].append(9)
assert grid == [[1], [2]], f"shallow copy leaked: {grid}"
check(detach_first_row, [[1], [2]], expect=[[1], [2]])
````
