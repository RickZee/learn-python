## Debug

This function has the bug from the lesson's Common mistake. Find it, say why it happens in one sentence, then fix it.


````python
def group_names(people: list[tuple[str, str]], cities: list[str]) -> dict[str, list[str]]:
    groups = dict.fromkeys(cities, [])
    for name, city in people:
        groups[city].append(name)
    return groups


check(
    group_names,
    [("Ada", "Paris"), ("Bea", "Rome")],
    ["Paris", "Rome"],
    expect={"Paris": ["Ada"], "Rome": ["Bea"]},
)
````
