## The idea

`enumerate` walks one list with an index. `zip` walks two lists and hands you pairs. Both are lazy until a loop or `list` pulls. Without `strict=True`, `zip` stops at the shorter list.

<details><summary>Explain more</summary>

A pair from `zip` is a tuple. `enumerate` starts at `0` unless you pass `start`.
</details>


## See it work

Pair each person with a room. The lists must be the same length, or the roster is wrong.


```python
def roster(names, rooms):
    # 1. strict refuses a leftover name or a leftover room.
    lines = []
    for name, room in zip(names, rooms, strict=True):
        # 2. One line per pair. zip already stopped the indexes lining up.
        lines.append(name + " in " + room)
    return lines


print(roster(["ada", "grace"], ["lab", "hall"]))
```

## Why it works this way

`zip` and `enumerate` are iterators. They store the inputs and a position, and build the next tuple only when the loop asks. A second loop over that same zip object does not start over. A zip of a million pairs is small until you pull. Deep dive: [The iteration protocol](../../internals/07-iteration/01-protocol-lesson.ipynb).

Stopping at the shorter list drops data and keeps going. `strict=True` raises `ValueError` when one side ends first — unequal inputs are usually a mistake, and the check is cheap next to a missing row. `enumerate` is the same lazy idea with a running counter instead of a second sequence. Pulling into `list(zip(...))` forces every pair into memory at once and keeps it.


```python
import sys

pairs = zip(range(10**6), range(10**6), strict=False)
print(sys.getsizeof(pairs) < 1000)
print(list(zip([1, 2], ["a"], strict=False)))
```

## When to use it

Use `zip` to walk lists in lockstep and `enumerate` when you need the index. Use `strict=True` whenever a short list would drop a row. Don't pair by `range(len(names))` and then index.


## Common mistake

Pairing unequal lists without `strict`. Grace has no room in the output. Nothing raises. The extra name is gone.


```python
names = ["ada", "grace"]
rooms = ["lab"]
print(list(zip(names, rooms, strict=False)))
```

`strict=True` makes the length mismatch the error. Use it when every name needs a room. Equal lists still pair.


```python
names = ["ada", "grace"]
rooms = ["lab", "hall"]
print(list(zip(names, rooms, strict=True)))
```

Unequal lists now raise. Read the last line first: `ValueError`. The arrow marks the `zip` that was asked to finish both sides.


```python
names = ["ada", "grace"]
rooms = ["lab"]
print(list(zip(names, rooms, strict=True)))
```
