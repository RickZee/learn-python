## The idea

A tuple groups values you treat as one thing, in a fixed order. You can unpack it into names. You cannot replace one item. Because the items cannot be swapped out, a tuple of hashable values can be a dict key.

<details><summary>Explain more</summary>

`left, right = pair` stores each item in order. `(10, 20)` is the clear form. A one-item tuple needs a comma: `(10,)`.
</details>


## See it work

Put two names into alphabetical order and return them as one value.


```python
def ordered(first, second):
    # 1. The earlier name goes first. Both results are new tuples.
    if first <= second:
        return (first, second)
    # 2. Otherwise swap by building the other order.
    return (second, first)


print(ordered("bea", "ada"))
print(ordered("ada", "ada"))
```

## Why it works this way

A tuple is a fixed block of pointers — length is part of the value, so there is no `append`. A dict finds a key by hashing it, then checking `==`. If that hash changed later, the next lookup would miss the entry. Numbers, strings, and tuples of those are safe keys; a list is not, and a tuple that holds a list is not. Hashing a tuple mixes the hashes of its items once at insert. Deep dive: [How dict and set work](../../internals/04-dicts/01-hash-table-lesson.ipynb). Fixed length also skips list overallocation, so millions of small key tuples cost less memory than the same rows as lists. Unpacking reads each pointer once; cost grows with length.


```python
pair = (1, 2)
table = {pair: "origin"}
print(hash((1, 2)) == hash((1, 2)))
print(table[(1, 2)])
```

```python
table = {}
key = [1, 2]
table[key] = "no"
```

## When to use it

Use a tuple for a fixed record or a two-part key. Use a list when the collection grows. A tuple does not freeze a list stored inside it — that inner list can still change.


## Common mistake

Building the updated tuple and not storing it. The expression makes a new tuple and throws it away. The name still points at the original pair.


```python
point = (1, 2)
(point[0] + 1, point[1])
print(point)
```

A tuple cannot be edited, so the only way to "change" it is to bind the name to a new one.


```python
point = (1, 2)
point = (point[0] + 1, point[1])
print(point)
```

Replacing an item raises. Read the last line first: `TypeError`. The arrow marks the assignment. Build a new tuple instead.


```python
pair = (1, 2)
pair[0] = 9
print(pair)
```

## Recap

- A tuple is a fixed group. Unpacking stores each item in a name.
- It can be a dict key when every item can. A list cannot.
- To change a field, bind the name to a new tuple.
