## Debug

This function has the bug from the lesson's Common mistake. Find it, say why it happens in one sentence, then fix it.

`roundtrip_text` should write and read with an explicit encoding.


```python
import tempfile
from pathlib import Path


def roundtrip_text(folder: str) -> str:
    path = Path(folder) / "demo.txt"
    path.write_text("café", encoding="utf-8")
    return path.read_text(encoding="latin-1")  # wrong codec


import tempfile

with tempfile.TemporaryDirectory() as folder:
    got = roundtrip_text(folder)
assert got == "café", "read with the same encoding you wrote"
```
