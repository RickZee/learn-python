## Debug

`decode_cafe` should decode UTF-8 bytes with UTF-8, not latin-1.


```python
def decode_cafe() -> str:
    data = "café".encode("utf-8")
    return data.decode("latin-1")  # wrong codec


assert decode_cafe() == "café"
```
