## Debug

This function has the bug from the lesson's Common mistake. Find it, say why it happens in one sentence, then fix it.

`choose` should use a cache on recursive calls. The planted bug wraps the function but never rebinds the name the body calls.


```python
import functools


def choose_hits(n: int, k: int) -> int:
    calls = {"n": 0}

    def choose(a: int, b: int) -> int:
        calls["n"] += 1
        if b == 0 or b == a:
            return 1
        return choose(a - 1, b - 1) + choose(a - 1, b)

    functools.cache(choose)  # BUG: never rebound to choose
    choose(n, k)
    return calls["n"]


from pylearn import check

# With a working cache rebind, choose(6, 3) needs far fewer than the raw 39 calls
check(choose_hits, 6, 3, expect=15)
```
