## Debug

This function has the bug from the lesson's Common mistake. Find it, say why it happens in one sentence, then fix it.

`tree_sum_stack` should walk every node with an explicit stack. The planted bug pops only the value and never pushes children.


```python
def tree_sum_stack(node: tuple) -> int:
    pending = [node]
    total = 0
    while pending:
        value = pending.pop()[0]  # drops children
        total += value
    return total


from pylearn import check

sample = (2, ((1, ()), (4, ((3, ()),))))
check(tree_sum_stack, sample, expect=10)
check(tree_sum_stack, (7, ()), expect=7)
```
