## Debug

This function has the bug from the lesson's Common mistake. Find it, say why it happens in one sentence, then fix it.

`tree_sum` should add every node value. The planted bug adds only `child[0]` and skips grandchildren.


```python
def tree_sum(node: tuple) -> int:
    value, children = node
    total = value
    for child in children:
        total += child[0]  # skips descendants
    return total


from pylearn import check

sample = (2, ((1, ()), (4, ((3, ()),))))
check(tree_sum, sample, expect=10)
check(tree_sum, (5, ()), expect=5)
```
