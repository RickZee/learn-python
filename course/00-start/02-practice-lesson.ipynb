## The idea

A function packages steps under a name. `def` starts it. `return` sends a value back. A call uses the name and parentheses. Type hints are labels for readers and tools. Python does not enforce them. Practice cells start with `raise NotImplementedError`. You replace that line. `check` prints ✓ when your return value matches.

<details><summary>Explain more</summary>

The indented lines under `def` are the body. The call `greet("world")` runs that body. `check(greet, "world", expect="hello world")` calls it for you and compares.
</details>


## See it work

Double a number and return it. The practice greeting is a different function.


```python
def double(n: int) -> int:
    # 1. Use the argument.
    # 2. Send the result back.
    return n * 2


print(double(4))
```

## Why it works this way

`def` compiles the body into a function object and stores it under the name. The body runs at the call, not at `def`. `return` hands a value back. `print` writes a line, and the call still ends as `None` if nothing was returned — so the example above writes `hello world` and then `None`.

Edit the body and forget to re-run the notebook cell, and later calls use the old object. A `.py` file does not: each launch reads, compiles, and runs from the top. [Source to bytecode](../../internals/10-bytecode/01-bytecode-lesson.ipynb) is the compile. [The import system](../../internals/09-imports/01-import-system-lesson.ipynb) is why importing runs the `def` once.

Read a traceback's last line first. `NotImplementedError` fails loudly on a stub. If `double` is undefined, a later call ends in `NameError`.


## When to use it

Use `return` when the caller needs the value; use `print` when a person should see a line. A function that only prints returns `None`, a poor hand-off to `check`.

Use a notebook to try one function; a `.py` file when the next run must not remember the last `def`. Raise `NotImplementedError` in a stub for a clear last line.


## Common mistake

The last line of the error names the problem: the function stops on purpose because the body is not written yet.


```python
def unfinished() -> str:
    raise NotImplementedError


unfinished()
```
