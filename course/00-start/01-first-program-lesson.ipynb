## The idea

A program is a file of instructions Python runs from top to bottom. `print` writes a line. A comment starts with `#` and is for the reader. When something goes wrong, read the last line of the error first: it names the problem.

<details><summary>Explain more</summary>

You run a file with `uv run python hello.py` from the folder that contains it. `input` reads a line the person types. A notebook would stop on `input()` and wait for a person. `ask.py` is a real file, and the cell below feeds it one line.
</details>

<details><summary>If you already write Java, JavaScript, or C#</summary>

A first program is a file you run. It does not need a class or a `public static` method.
</details>


## See it work

Run the real file `hello.py` the way you would from a terminal.


```python
import subprocess
import sys

print(subprocess.run([sys.executable, "hello.py"], capture_output=True, text=True).stdout)
```

## input

`ask.py` calls `input()`, which reads one line. The notebook feeds that line in, so you do not have to type.


```python
import subprocess
import sys

completed = subprocess.run(
    [sys.executable, "ask.py"],
    input="ada\n",
    capture_output=True,
    text=True,
)
print(completed.stdout)
```

## Why it works this way

A `.py` file is compiled to a code object, then run top to bottom. A later line cannot use a name an earlier line has not assigned. Nothing is kept for the next launch. [Source to bytecode](../../internals/10-bytecode/01-bytecode-lesson.ipynb) is the compile. [The import system](../../internals/09-imports/01-import-system-lesson.ipynb) is the same start when the file is imported: one module, run once.

A notebook is one long-lived process. The cell above sets `runs` to 1, prints it, then adds 1 — a later cell sees 2. Re-run and `runs = 1` starts over. Skip it and run only `runs = runs + 1`, and the error ends in `NameError`. `hello.py` avoids that: every launch starts at the top.


```python
runs = 1
print("first", runs)
runs = runs + 1
print("second", runs)
```

## When to use it

Run a `.py` file when the program should start empty every time. Use a notebook when you want to keep a name and try the next line. If a cell only works because an earlier cell already ran, a line is missing.

Read a traceback from the bottom — the last line names the error.


## Common mistake

The last line of the error names the problem: that name was never given a value.


```python
label = "ada"
del label
print(label)  # noqa: F821
```
