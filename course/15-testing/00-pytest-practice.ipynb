## Debug

This function has the bug from the lesson's Common mistake. Find it, say why it happens in one sentence, then fix it.

`isolated_label` should not depend on a leftover file from another test.


```python
from pathlib import Path


def isolated_label() -> str:
    shared = Path("exercise/_shared_out.txt")
    shared.write_text("stale", encoding="utf-8")  # planted: shared leftover path
    return shared.read_text(encoding="utf-8")


try:
    assert isolated_label() == "fresh", "return a local label; do not leave a shared file"
finally:
    Path("exercise/_shared_out.txt").unlink(missing_ok=True)
```
