## The idea

A good test arranges inputs, acts by calling the code, and asserts on the result. Edge cases matter: empty lists, zeros, negatives. Determinism matters: same inputs, same result, no hidden clock or network.

<details><summary>Explain more</summary>

`assert is_even(0) is True` locks a decision about zero. Without it, a later change can silently flip behaviour.
</details>

Write the failing case first when you can. Watching it go red, then green, proves the test can fail.


## See it work

Cover even numbers including zero.


```python
def is_even(n: int) -> bool:
    # 1. Zero is even.
    return n % 2 == 0


def test_even() -> None:
    assert is_even(2) is True
    assert is_even(3) is False
    assert is_even(0) is True


test_even()
print("passed")
```

## Why it works this way

Tests buy you a short feedback loop. They document intended behaviour next to the code and fail when that behaviour drifts. Arrange-act-assert keeps each test readable: set up, call, check. Edge cases are where bugs hide, so empty inputs and zeros belong in the suite early. Name tests after behaviour, not after implementation details. When a test needs a paragraph of setup, extract a helper or fixture instead of copying blocks.

Order independence is part of isolation. If test B needs test A to run first, you have shared state, not a suite. Prefer pure functions and fixtures over mutating globals. When a test must touch time or IO, inject fakes (next lesson) instead of hoping the environment stays still.

Cost should stay low enough that you run the suite often. A slow suite becomes a suite you skip. The design reason for many small tests is localization: a failure points at one behaviour.

Evidence: a focused `test_even` that includes zero, and the habit of calling the test function in the notebook before you move it into a file for pytest collection. Re-run the suite after every failing fix to confirm you only changed what you meant.

Regression tests earn their keep the first time they catch a change you did not mean. Characterise current behaviour when you inherit code without docs, then improve the behaviour behind the green bar. Avoid testing private helpers unless they hold tricky logic; prefer asserting through the public function.

Flaky tests train the team to ignore red builds. Determinism is therefore a social property as well as a technical one. Seed randomness, freeze time, and isolate files. When a test needs a paragraph of narrative comments, the API under test is probably too awkward—improve the API or add a seam.


```python
def is_even(n: int) -> bool:
    return n % 2 == 0


def evens(nums: list[int]) -> list[int]:
    return [n for n in nums if is_even(n)]


def test_evens() -> None:
    assert evens([1, 2, 3, 4]) == [2, 4]
    assert evens([]) == []
    assert evens([1, 3]) == []


test_evens()
print("passed")
```

## When to use it

Write tests before or with behaviour changes. Cover edges. Keep them fast and deterministic. Don't assert on irrelevant formatting. Don't share writable module globals across tests. Add a failing test before you fix a bug so the fix cannot silently disappear later.


## Common mistake

Testing only the happy path and missing empty input.


```python
def evens(nums: list[int]) -> list[int]:
    return [n for n in nums if n % 2 == 0]


assert evens([2, 4]) == [2, 4]
print("empty list never checked")
```

Add `assert evens([]) == []` so the empty case stays locked.


```python
def evens(nums: list[int]) -> list[int]:
    return [n for n in nums if n % 2 == 0]


assert evens([2, 4]) == [2, 4]
assert evens([]) == []
print("empty covered")
```
