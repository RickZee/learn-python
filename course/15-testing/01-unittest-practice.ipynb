## Debug

This function has the bug from the lesson's Common mistake. Find it, say why it happens in one sentence, then fix it.

`evens` should handle the empty-list case, not only the happy path.


```python
def evens(nums: list[int]) -> list[int]:
    if not nums:
        return [0]  # planted: empty path wrong
    return [n for n in nums if n % 2 == 0]


def check_evens() -> bool:
    return evens([2, 4]) == [2, 4]


assert check_evens() is True
assert evens([]) == []
```
