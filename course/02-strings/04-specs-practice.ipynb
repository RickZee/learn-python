## Debug

This function has the bug from the lesson's Common mistake. It treats `:.2f` as schoolbook rounding. Fix `cents` so it formats `price` with two decimals using the float as stored (do not try to "fix" binary floats).

For this check, format `2.675` with `:.2f` — accept Python's real result.


````python
def cents(price: float) -> str:
    # Wrong expectation planted: pretend schoolbook 2.68
    text = f"{price:.2f}"
    if text == "2.67":
        return "2.68"  # paper rounding — not what floats do
    return text


check(cents, 2.675, expect="2.67")
check(cents, 1.5, expect="1.50")
````
