## The idea

An f-string inserts values into text. Write `f` before the quotes and an expression inside `{}`. Python evaluates that expression when the line runs. Without `f`, the braces are ordinary characters.

<details><summary>Explain more</summary>

`{path}` inserts the text of the value. `{path!r}` inserts the repr, quotes included, so a trailing space stays visible. `{size + 1}` adds, then inserts the result.
</details>


## See it work

A one-line label for a file. The repr keeps a trailing space visible.


```python
def file_label(path, size):
    # 1. The f makes the braces expressions, evaluated when this line runs.
    # 2. !r uses repr, so quotes show a space that a plain insertion would hide.
    return f"{path!r} ({size} bytes)"


print(file_label("notes.txt", 12))
print(file_label("notes.txt ", 12))
```

</details>


## Why it works this way

An f-string is compiled with the surrounding code. When the line runs, Python evaluates each braced expression, converts the result, and builds the label. A missing name fails on this line, not at `def` time. `size + 1` runs as ordinary code; the text receives the result.

Without `f`, braces are not special — a template stored as data stays data until you opt in. `!r` asks for `repr` (quotes and spaces); `!s` is the plain conversion, and writing neither means `!s`.

A braced string that is the whole f-string can be the same object as the input — CPython detail: a copy would waste. Add characters around it and Python allocates. `+` between a string and a number raises; the f-string makes the conversion explicit. Deep dive: [Numbers and strings](../../internals/05-numbers/01-numbers-strings-lesson.ipynb).


```python
path = "notes.txt"
print(f"{path}" is path)
print(f"{path} file" is path)
```

## When to use it

Use an f-string when a value is a number or needs `repr`. Use `+` for two or three pieces already text. Keep a data template as a plain string. Do not build a shell command by inserting values — pass them separately.


## Common mistake

Forgetting `f`. The braces look like placeholders, and the line prints them. No name was looked up, so this does not raise.


```python
path = "notes.txt"
size = 12
label = "{path} is {size} bytes"
print(label)
```

The `f` is what evaluates the braces. Without it they are characters, so the names appear instead of the path and the size.


```python
path = "notes.txt"
size = 12
label = f"{path} is {size} bytes"
print(label)
```
