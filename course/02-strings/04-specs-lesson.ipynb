## The idea

Inside an f-string, a colon starts a format spec. It describes the rendered text, not a new number. `:.2f` keeps two decimal digits. `:,` groups thousands. `:>8` pads on the left to a width of at least 8. Combine them as `{price:,.2f}`.

<details><summary>Explain more</summary>

`<` pads on the right. The width is a minimum. A longer value is not cut.
</details>


## See it work

One row of a score table: the name on the left, the points on the right.


```python
def score_line(name, points):
    # 1. Name, left side of a 12-character field.
    # 2. Points, one decimal, right side of a 6-character field.
    return f"{name:<12}{points:>6.1f}"


print(score_line("Ada", 9.5))
print(repr(score_line("Grace", 10)))
```

## Why it works this way

A format spec describes rendered text; it does not change the number. `.2f` rounds the float that is stored, not the digits you typed. A float is a binary fraction — the literal `2.675` sits a little under the decimal, so rounding to two places produces `2.67`. [Numbers and strings](../../internals/05-numbers/01-numbers-strings-lesson.ipynb) shows the fraction.

`,` only inserts a separator into the text. `>` pads on the left until the text is at least that wide — a minimum, because cutting a longer rendering would show a different quantity. Exact tenths belong in an integer of cents; `//` and `%` never pass through the binary fraction that turned `2.675` into `2.67`. Each format call builds one new string sized to the rendered width.


```python
print(f"{2.675:.2f}")
print(f"{2.675:.20f}")
print(repr(f"{123456789:>8}"))
```

## When to use it

Use a spec to line up a column or fix decimals. For money, store cents and format with `//` and `%`. Use `>` or `<` for a minimum width. When text must be cut, slice it — the spec will not.


## Common mistake

Expecting `:.2f` to round the decimal you typed. `2.675` and `1.005` are already inexact before the spec runs, and both print short of the schoolbook result.


```python
def shown(price):
    return f"{price:.2f}"


print(shown(2.675))
print(shown(1.005))
```

Store cents. `//` and `%` split an exact integer. `02d` prints the remainder with a leading zero, so 268 cents is `2.68` and 5 cents is `0.05`.


```python
def shown_cents(cents):
    whole = cents // 100
    frac = cents % 100
    return f"{whole}.{frac:02d}"


print(shown_cents(267))
print(shown_cents(268))
print(shown_cents(5))
```
