## Core

Write `tracked` so the yielded list records `"open"`, then whatever the body appends, then `"close"` even if the body raises.

<details><summary>Hint: the idea</summary>

Yield a list. Append close in `finally`.

</details>

<details><summary>Hint: the approach</summary>

`@contextmanager` plus `try`/`yield`/`finally`.

</details>

<details><summary>Hint: near the code</summary>

`events = ["open"]` then `yield events` inside `try`, and `events.append("close")` in `finally`.

</details>


```python
@contextmanager
def tracked():
    raise NotImplementedError


with tracked() as events:
    events.append("inside")
assert events == ["open", "inside", "close"], events

try:
    with tracked() as events:
        events.append("inside")
        raise RuntimeError("boom")
except RuntimeError:
    pass
assert events == ["open", "inside", "close"], events
check(lambda: events[-1], expect="close")
```

## Stretch

Yield a temporary working dict that is cleared in `finally`. After the `with`, the dict should be empty even if the body stored keys.

<details><summary>Hint: the idea</summary>

Cleanup must clear the same dict object that was yielded.

</details>

<details><summary>Hint: the approach</summary>

`data.clear()` in `finally`.

</details>

<details><summary>Hint: near the code</summary>

`data = {}` / `yield data` / `data.clear()`.

</details>


```python
@contextmanager
def scratch():
    raise NotImplementedError


with scratch() as data:
    data["x"] = 1
    assert data["x"] == 1
assert data == {}
check(lambda: data, expect={})
```
