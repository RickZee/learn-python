## Core

Return the log from a `Door` context manager.

<details><summary>Hint: the idea</summary>

`__enter__` starts the log. The block appends `inside`. `__exit__` appends `close`.
</details>

<details><summary>Hint: the approach</summary>

Return `self` from `__enter__` so `as` has the Door.
</details>

<details><summary>Hint: near the code</summary>

`self.log = ["open"]` then append in the block and in `__exit__`.
</details>


```python
class Door:
    def __enter__(self):
        raise NotImplementedError

    def __exit__(self, exc_type, exc, tb):
        raise NotImplementedError


def visit() -> list[str]:
    with Door() as entered:
        entered.log.append("inside")
        return entered.log


check(visit, expect=["open", "inside", "close"])
```
