## Core

Return `bag[index]` for a `Bag` you can index.

<details><summary>Hint: the idea</summary>

`__getitem__` receives the index.
</details>

<details><summary>Hint: the approach</summary>

Return `self.items[index]`.
</details>

<details><summary>Hint: near the code</summary>

`def __getitem__(self, index: int) -> str`
</details>


```python
class Bag:
    def __init__(self, items: list[str]) -> None:
        self.items = items

    def __getitem__(self, index: int) -> str:
        raise NotImplementedError


def at(items: list[str], index: int) -> str:
    return Bag(items)[index]


check(at, ["a", "b"], 0, expect="a")
check(at, ["a", "b"], 1, expect="b")
```

## Debug

This function has the bug from the lesson's Common mistake. Find it, say why it happens in one sentence, then fix it.

`unique_boxes` should put equal Box values in a set. The planted bug defines `__eq__` without `__hash__`.


```python
class Box:
    def __init__(self, value: int) -> None:
        self.value = value

    def __eq__(self, other) -> bool:
        return isinstance(other, Box) and self.value == other.value


def unique_boxes() -> int:
    try:
        return len({Box(1), Box(1)})
    except TypeError:
        return -1  # unhashable with __eq__ but no __hash__


assert unique_boxes() == 1, "define __hash__ alongside __eq__"
```

## Stretch

Return whether the bag contains the item.


```python
class Pocket:
    def __init__(self, items: list[str]) -> None:
        self.items = items

    def __contains__(self, item: object) -> bool:
        raise NotImplementedError


def holds(items: list[str], item: str) -> bool:
    return item in Pocket(items)


check(holds, ["a"], "a", expect=True)
check(holds, ["a"], "z", expect=False)
```
