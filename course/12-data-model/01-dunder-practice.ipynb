## Core

Return `bag[index]` for a `Bag` you can index.

<details><summary>Hint: the idea</summary>

`__getitem__` receives the index.
</details>

<details><summary>Hint: the approach</summary>

Return `self.items[index]`.
</details>

<details><summary>Hint: near the code</summary>

`def __getitem__(self, index: int) -> str`
</details>


```python
class Bag:
    def __init__(self, items: list[str]) -> None:
        self.items = items

    def __getitem__(self, index: int) -> str:
        raise NotImplementedError


def at(items: list[str], index: int) -> str:
    return Bag(items)[index]


check(at, ["a", "b"], 0, expect="a")
check(at, ["a", "b"], 1, expect="b")
```

## Stretch

Return whether the bag contains the item.


```python
class Pocket:
    def __init__(self, items: list[str]) -> None:
        self.items = items

    def __contains__(self, item: object) -> bool:
        raise NotImplementedError


def holds(items: list[str], item: str) -> bool:
    return item in Pocket(items)


check(holds, ["a"], "a", expect=True)
check(holds, ["a"], "z", expect=False)
```
