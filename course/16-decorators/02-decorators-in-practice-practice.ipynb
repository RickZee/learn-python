## Core

Write a class decorator `Counted` with `__call__` that counts calls on `self.calls`.

<details><summary>Hint: the idea</summary>

Store the function in `__init__`. Increment in `__call__`, then call the stored function.
</details>

<details><summary>Hint: the shape</summary>

`functools.update_wrapper(self, func)` keeps the name.
</details>

<details><summary>Hint: near the code</summary>

```python
class Counted:
    def __init__(self, func):
        functools.update_wrapper(self, func)
        self.func = ___
        self.calls = 0
    def __call__(self, *args, **kwargs):
        self.calls += 1
        return self.func(*args, **kwargs)
```
</details>


````python
class Counted:
    def __init__(self, func):
        raise NotImplementedError

    def __call__(self, *args, **kwargs):
        raise NotImplementedError


@Counted
def add_one(n: int) -> int:
    return n + 1


assert add_one(3) == 4
assert add_one.calls == 1
assert add_one(4) == 5
assert add_one.calls == 2
print("✓ Counted")
````

## Debug

This function has the bug from the lesson's Common mistake. Find it, say why it happens in one sentence, then fix it.

`try_hello` should return `"hi Ada"`. The planted wrapper forgets `self` in its signature.


````python
def as_method(func):
    def wrapper(name: str) -> str:
        return func(name)  # planted: drops self

    return wrapper


class Greeter:
    def __init__(self, label: str) -> None:
        self.label = label

    @as_method
    def hello(self, name: str) -> str:
        return self.label + " " + name


def try_hello() -> str:
    try:
        return Greeter("hi").hello("Ada")
    except TypeError:
        return "broken"


assert try_hello() == "hi Ada", "forward self in the wrapper"
````
