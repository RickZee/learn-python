## Debug

This function has the bug from the lesson's Common mistake. Find it, say why it happens in one sentence, then fix it.

`double` should return the wrapper. The planted bug returns the original function unchanged.


```python
def double(func):
    def wrapper(n: int) -> int:
        return func(n) * 2

    return func  # planted: returns original, no enhancement


@double
def three() -> int:
    return 3


assert three() == 6, "return the wrapper from the decorator"
```
