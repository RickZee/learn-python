# Machine learning

The point (0, 0) matches the closer of two labeled points.


```python
def nearest(point, labeled):
    best_label = None
    best_distance = None
    for (x, y), label in labeled:
        distance = (point[0] - x) ** 2 + (point[1] - y) ** 2
        if best_distance is None or distance < best_distance:
            best_label = label
            best_distance = distance
    return best_label


labeled = [((0, 0), "a"), ((10, 10), "b")]
print(nearest((1, 1), labeled))
```
