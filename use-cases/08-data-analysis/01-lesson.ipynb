# Data analysis and visualization

Two cities, three rows, one average each.


```python
from collections import defaultdict

rows = [
    {"city": "rome", "score": 10},
    {"city": "rome", "score": 6},
    {"city": "oslo", "score": 8},
]
buckets = defaultdict(list)
for row in rows:
    buckets[row["city"]].append(row["score"])
summary = {city: sum(scores) / len(scores) for city, scores in buckets.items()}
print(summary)
```
