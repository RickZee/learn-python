# Data engineering

Three raw records go in. The blank amount is dropped. The rest become numbers.


```python
raw = [
    {"id": "1", "amount": "10"},
    {"id": "2", "amount": ""},
    {"id": "3", "amount": "4"},
]
clean = []
for row in raw:
    if row["amount"] == "":
        continue
    clean.append({"id": row["id"], "amount": int(row["amount"])})
print(clean)
```
