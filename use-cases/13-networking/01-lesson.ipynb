# Networking

A URL splits, and a private address is recognized.


```python
import ipaddress
from urllib.parse import urlparse

parts = urlparse("https://example.com/docs")
print(parts.scheme, parts.netloc, parts.path)
print(ipaddress.ip_address("10.0.0.1").is_private)
print(ipaddress.ip_address("8.8.8.8").is_private)
```
