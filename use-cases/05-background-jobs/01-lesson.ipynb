# Background jobs

A worker retries each task once and collects the failures.


```python
def run_tasks(tasks: list[tuple[str, bool]]) -> list[str]:
    failed = []
    for name, ok in tasks:
        if ok:
            continue
        second_try = ok
        if not second_try:
            failed.append(name)
    return failed


print(run_tasks([("mail", True), ("resize", False)]))
```
