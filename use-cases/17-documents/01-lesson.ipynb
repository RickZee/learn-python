# Documents and office files

Two records survive a trip through CSV.


```python
import csv
import io

records = [{"name": "ada", "score": "10"}, {"name": "grace", "score": "8"}]
buffer = io.StringIO()
writer = csv.DictWriter(buffer, fieldnames=["name", "score"])
writer.writeheader()
writer.writerows(records)
buffer.seek(0)
print(list(csv.DictReader(buffer)))
```
