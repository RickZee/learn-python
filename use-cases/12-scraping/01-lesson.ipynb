# Web scraping

Two anchor tags yield two titles.


```python
from html.parser import HTMLParser


class LinkText(HTMLParser):
    def __init__(self):
        super().__init__()
        self._capture = False
        self.found = []

    def handle_starttag(self, tag, attrs):
        self._capture = tag == "a"

    def handle_endtag(self, tag):
        if tag == "a":
            self._capture = False

    def handle_data(self, data):
        if self._capture:
            self.found.append(data.strip())


parser = LinkText()
parser.feed("<p><a>Ada</a> <a>Grace</a></p>")
print(parser.found)
```
