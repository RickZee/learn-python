# Desktop, terminal, and mobile UI

A menu string and a function that applies choice 1 or 2.


```python
def menu() -> str:
    return "1) list\n2) quit"


def apply(choice: str) -> str:
    if choice == "1":
        return "listed"
    if choice == "2":
        return "bye"
    return "unknown"


print(menu())
print(apply("1"), apply("9"))
```
