# APIs and services

One function checks a payload and returns an error status or the greeting.


```python
def greet(payload: dict) -> tuple[int, dict]:
    if "name" not in payload or not payload["name"]:
        return 400, {"error": "name is required"}
    return 200, {"message": f"hello {payload['name']}"}


print(greet({"name": "ada"}))
print(greet({}))
```
