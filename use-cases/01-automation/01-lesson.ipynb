# Automation and scripting

A CSV string becomes rows, the score column is kept, and the rows are printed.


```python
import csv
import io

raw = "name,score\nada,10\ngrace,8\n"
rows = list(csv.DictReader(io.StringIO(raw)))
kept = [{"name": row["name"], "score": int(row["score"])} for row in rows]
print(kept)
```
