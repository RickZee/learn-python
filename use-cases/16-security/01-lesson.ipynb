# Security

Two equal digests compare as true. A third does not.


```python
import hashlib
import secrets


def digest(text: str) -> str:
    return hashlib.sha256(text.encode()).hexdigest()


left = digest("correct")
right = digest("correct")
other = digest("other")
print(secrets.compare_digest(left, right))
print(secrets.compare_digest(left, other))
```
