## How to spend a timed round

1. Restate the problem. Confirm inputs, outputs, constraints, duplicates, empty input, and whether the data is sorted.
2. Ask about edge cases and the return format.
3. Say a brute-force idea and its complexity, even when you discard it immediately.
4. Propose the better approach, name the data structure, and state the time and space you expect.
5. Code while narrating the decisions that matter.
6. Walk one or two examples, including an edge case.
7. State the final complexity and one trade-off.

Reviewers grade the analysis and the trade-offs they can hear. A memorized solution with no narration is weaker.


## Patterns, in the order to learn them

- **Hash map, set, Counter.** Complements, frequencies, grouping, values already seen.
- **Two pointers.** Sorted arrays, pairs, partitioning, palindromes.
- **Sliding window.** A contiguous subarray or substring with a condition on the sum, the unique characters, or a min or max.
- **Binary search.** A sorted array, a rotated array, or a search over the answer space.
- **Stack, including a monotonic stack.** Parentheses, next greater element, histograms.
- **Intervals.** Sort, then merge or sweep.
- **Trees and graphs.** DFS, BFS, connected components, cycles, level order.
- **Heaps.** Top-k, merge k lists, a running median.
- **Linked lists.** Reverse, merge, cycle, dummy head.
- **Basic dynamic programming.** One dimension first: climbing stairs, house robber, coin change. Then a simple grid or string.

Pattern recognition plus clean Python beats a random problem list.

## Cost table for Python built-ins

Average case unless noted. Deep dive: [How list works](../internals/03-lists/01-dynamic-array-lesson.ipynb), [How dict and set work](../internals/04-dicts/01-hash-table-lesson.ipynb).

| Structure / op | Time | Extra memory | Notes |
| --- | --- | --- | --- |
| `list.append` | amortized O(1) | O(1) | over-allocates |
| `list.insert(0)` / `pop(0)` | O(n) | O(1) | shifts elements |
| `x in list` | O(n) | O(1) | scan |
| `dict` / `set` lookup, add | average O(1) | O(n) | keys hashable |
| `deque.append` / `appendleft` / `pop` / `popleft` | O(1) | O(n) | both ends |
| `heapq.heappush` / `heappop` | O(log n) | O(1) | min-heap on a list |
| `heapq.heapify` | O(n) | O(1) | better than n pushes |
| `str` concat in a loop (`+=`) | O(n²) total | O(n) | use `"".join` |
| `bisect_left` on sorted list | O(log n) | O(1) | `insort` is O(n) |

Usual default: reach for `dict`/`set` for membership, `deque` for queues, `heapq` for top-k, avoid `insert(0)` on lists.

## Idioms that save time

`deque` is the queue. `list.pop(0)` walks the whole list. `heapq` is a min-heap; push the negation when you want a max-heap. A dummy node gives linked-list code one tail to move.

Prefer `enumerate`, `zip`, and names like `left`, `right`, `freq`, and `window_sum`. Write the readable version first.


```python
import heapq
from collections import Counter, defaultdict, deque

nums = [1, 2, 2, 3]
print(Counter(nums))

words = ["eat", "tea", "tan"]
groups = defaultdict(list)
for word in words:
    groups[tuple(sorted(word))].append(word)
print(dict(groups))

q = deque([1, 2, 3])
q.append(4)
print(q.popleft())

heap = []
for value in (3, 1, 2):
    heapq.heappush(heap, -value)
print(-heapq.heappop(heap))
```

## Pitfalls

`[[0] * n] * m` repeats one row. Build each row on its own.

A default list is created once and shared by later calls. Use `None`, then create the list inside the function.

`list.pop(0)` and `list.insert(0, x)` are linear. Use a `deque`.

`heapq` pops the smallest value. Negate values, or push tuples, when you need the largest or a tie broken on purpose.

Write the midpoint as `low + (high - low) // 2`. In Python `(low + high) // 2` does not overflow, and the first form is still the safer habit in other languages.

Deleting from a dict while iterating it raises `RuntimeError`. Iterate over a copy of the keys.

`is` asks whether two names are the same object. `==` asks whether the values are equal. Use `is` for `None`.

Build a string with a list and `"".join`. Repeated `+=` on a string copies it each time.


```python
def shared_default(lst=[]):  # noqa: B006
    lst.append(1)
    return lst


print(shared_default(), shared_default())

rows, cols = 2, 3
aliased = [[0] * cols] * rows
aliased[0][0] = 1
print(aliased)

separate = [[0] * cols for _ in range(rows)]
separate[0][0] = 1
print(separate)

low, high = 0, 10
print(low + (high - low) // 2)

items = {"a": 1, "b": 2}
try:
    for key in items:
        del items[key]
except RuntimeError as error:
    print(type(error).__name__)

items = {"a": 1, "b": 2}
for key in list(items):
    if key == "a":
        del items[key]
print(items)

print([] == [], [] is [])  # noqa: F632
print("".join(["a", "b", "c"]))
```

## Reference solutions

These are the five walkthroughs. Use them after a real attempt. Each cell states the approach, then checks the examples.

**Longest substring.** Expand `right`. When the character was already seen inside the window, move `left` to just after that earlier index. Track the best length. Time O(n). Space O(min(n, alphabet)).

**3Sum.** Sort. Fix one index and run two pointers on the rest. Skip a duplicate at the fixed index and after every match. Time O(n²). Extra space O(1) besides the output.

**Merge intervals.** Sort by start. If the next start is within the last merged end, extend that end. Otherwise append. Touching endpoints overlap. Time O(n log n). Space O(n).

**Number of islands.** Flood-fill each unseen land cell and count the floods. Mark visited land as water. Time O(rows × cols). The recursive version uses that much stack in the worst case. A `deque` BFS avoids that depth.

**LRU cache.** `OrderedDict` keeps insertion order and moves a key to the end on use. `popitem(last=False)` drops the oldest. `get` and `put` are O(1) on average. A hash map plus a doubly linked list is the version to describe when the prompt wants the structure built by hand.
