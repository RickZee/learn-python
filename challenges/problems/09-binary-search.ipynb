# 09. Binary Search

**Difficulty:** Easy. **Target:** O(log n) time, O(1) extra space. **Time box:** 15.

You are given a sorted list of distinct integers `nums` and a `target`. Return the index of `target`, or `-1` if it is missing.

Lookup search is the base move for many sorted-array problems.

**Constraints.** `nums` has 0 to 10_000 distinct integers in increasing order.


## Examples

- `[-1,0,3,5,9,12]`, `9` → `4`.
- `[-1,0,3,5,9,12]`, `2` → `-1`.
- `[5]`, `5` → `0`.
- `[]`, `1` → `-1`.
