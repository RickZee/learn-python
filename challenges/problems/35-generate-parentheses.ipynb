# 35. Generate Parentheses

**Difficulty:** Medium. **Target:** Catalan-number many strings. **Time box:** 25 minutes.

Generate every string of `n` pairs of correctly matched parentheses. For `n = 3` that is the Catalan-number many strings such as `((()))` and `(()())`.

Editors and parsers enumerate the same well-formed shapes. Only valid nestings count; every open needs a matching close.

**Constraints.** `n` is between 1 and 6 in the checks.


## Examples

- `1` → `["()"]`.
- `2` → `["(())","()()"]` (any order).
- `3` → `["((()))","(()())","(())()","()(())","()()()"]`.
