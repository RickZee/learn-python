```python
import random


def min_distance(*args, **kwargs):
    raise NotImplementedError


def _brute_edit(a, b):
    memo = {}

    def go(i, j):
        if i == len(a):
            return len(b) - j
        if j == len(b):
            return len(a) - i
        key = (i, j)
        if key in memo:
            return memo[key]
        if a[i] == b[j]:
            memo[key] = go(i + 1, j + 1)
        else:
            memo[key] = 1 + min(go(i + 1, j), go(i, j + 1), go(i + 1, j + 1))
        return memo[key]

    return go(0, 0)


assert min_distance("horse", "ros") == 3
assert min_distance("intention", "execution") == 5
assert min_distance("", "") == 0
assert min_distance("a", "") == 1
assert min_distance("", "a") == 1
assert min_distance("a", "a") == 0
assert min_distance("ab", "ac") == 1
assert min_distance("plasma", "altruism") == 6
rng = random.Random(57)
for _ in range(40):
    a = "".join(rng.choice("abc") for _ in range(rng.randint(0, 6)))
    b = "".join(rng.choice("abc") for _ in range(rng.randint(0, 6)))
    assert min_distance(a, b) == _brute_edit(a, b)
print("all checks passed")
```
