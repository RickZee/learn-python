# 44. Sliding Window Maximum

**Difficulty:** Hard. **Target:** O(n) time, O(k) space. **Time box:** 30 minutes.

Return the maximum value in every contiguous window of size `k` as the window slides from left to right across `nums`.

Streaming monitors and max-filters use the same sliding peak.

**Constraints.** `nums` has length at least `k`. Lengths stay under 20 in checks.


## Examples

- `[1,3,-1,-3,5,3,6,7]`, `k = 3` → `[3,3,5,5,6,7]`.
- `[1]`, `k = 1` → `[1]`.
- `[9,8,7,6]`, `k = 2` → `[9,8,7]`.
- `[1,2,3]`, `k = 3` → `[3]`.
