```python
import random


def max_sliding_window(*args, **kwargs):
    raise NotImplementedError


def _brute_window_max(nums, k):
    if not nums:
        return []
    return [max(nums[i : i + k]) for i in range(len(nums) - k + 1)]


assert max_sliding_window([1, 3, -1, -3, 5, 3, 6, 7], 3) == [3, 3, 5, 5, 6, 7]
assert max_sliding_window([1], 1) == [1]
assert max_sliding_window([9, 8, 7], 2) == [9, 8]
assert max_sliding_window([1, 2], 2) == [2]
assert max_sliding_window([7, 2, 4], 2) == [7, 4]
assert max_sliding_window([1, 3, 1, 2, 0, 5], 3) == [3, 3, 2, 5]
assert max_sliding_window([1, 1, 1], 1) == [1, 1, 1]
assert max_sliding_window([4, 3, 2, 1], 4) == [4]
rng = random.Random(44)
for _ in range(40):
    n = rng.randint(1, 12)
    nums = [rng.randint(-20, 20) for _ in range(n)]
    k = rng.randint(1, n)
    assert max_sliding_window(nums, k) == _brute_window_max(nums, k)
print("all checks passed")
```
