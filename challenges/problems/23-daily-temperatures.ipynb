# 23. Daily Temperatures

**Difficulty:** Medium. **Target:** O(n) time, O(n) space. **Time box:** 25 minutes.

You are given daily temperatures `temperatures[i]`. For each day, return how many days you wait until a warmer temperature. If none exists, that entry is 0.

Monotonic stacks turn "next greater" scans into linear time.

**Constraints.** `temperatures` has 0 to 10_000 integers in a typical weather range.


## Examples

- `[73,74,75,71,69,72,76,73]` → `[1,1,4,2,1,1,0,0]`.
- `[30,40,50,60]` → `[1,1,1,0]`.
- `[30,30]` → `[0,0]`.
- `[]` → `[]`.
