# 13. Reverse Linked List

**Difficulty:** Easy. **Target:** O(n) time, O(1) extra space. **Time box:** 20 minutes.

You are given the head of a singly linked list. Reverse the links and return the new head. An empty list stays empty.

Reversing a chain of nodes is the base move for many list tricks.

**Constraints.** The list has 0 to 5000 nodes.


## Examples

- `[1,2,3]` → `[3,2,1]`.
- `[]` → `[]`.
- `[1]` → `[1]`.
- `[1,2]` → `[2,1]`.

</details>

````python
class ListNode:
    def __init__(self, val: int = 0, next: "ListNode | None" = None) -> None:
        self.val = val
        self.next = next


def nodes(values: list[int]) -> ListNode | None:
    dummy = ListNode()
    tail = dummy
    for value in values:
        tail.next = ListNode(value)
        tail = tail.next
    return dummy.next


def values(head: ListNode | None) -> list[int]:
    out: list[int] = []
    while head is not None:
        out.append(head.val)
        head = head.next
    return out


def reverse_list(head: ListNode | None) -> ListNode | None:
    raise NotImplementedError


from pylearn.challenges import cross_check

assert values(reverse_list(nodes([1, 2, 3]))) == [3, 2, 1]
assert values(reverse_list(nodes([]))) == []
assert values(reverse_list(nodes([1]))) == [1]
assert values(reverse_list(nodes([1, 2]))) == [2, 1]
assert values(reverse_list(nodes([5, 4, 3, 2, 1]))) == [1, 2, 3, 4, 5]
assert values(reverse_list(nodes([0]))) == [0]
assert values(reverse_list(nodes([1, 1, 1]))) == [1, 1, 1]
assert values(reverse_list(nodes([9, 8]))) == [8, 9]
cross_check("13-reverse-linked-list", lambda vals: values(reverse_list(nodes(vals))))
print("all checks passed")
````
