# 21. Maximum Subarray

**Difficulty:** Medium. **Target:** O(n) time, O(1) extra space. **Time box:** 20 minutes.

You are given a list of integers `nums` (at least one element). Return the largest sum of any contiguous subarray.

Kadane's algorithm is the standard linear scan for this.

 Empty subarrays are not allowed; the answer is always the sum of at least one element.

**Constraints.** `nums` has 1 to 10_000 integers.


## Examples

- `[-2,1,-3,4,-1,2,1,-5,4]` → `6` (`[4,-1,2,1]`).
- `[1]` → `1`.
- `[5,4,-1,7,8]` → `23`.
- `[-1]` → `-1`.
