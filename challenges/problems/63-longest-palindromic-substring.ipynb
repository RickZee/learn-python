```python
import random


def longest_palindrome(*args, **kwargs):
    raise NotImplementedError


def _brute_palindromes(s):
    best = ""
    for i in range(len(s)):
        for j in range(i, len(s)):
            sub = s[i : j + 1]
            if sub == sub[::-1] and len(sub) > len(best):
                best = sub
    return best


assert longest_palindrome("babad") in {"bab", "aba"}
assert longest_palindrome("cbbd") == "bb"
assert longest_palindrome("a") == "a"
assert longest_palindrome("ac") in {"a", "c"}
assert longest_palindrome("aaa") == "aaa"
assert longest_palindrome("forgeeksskeegfor") == "geeksskeeg"
assert longest_palindrome("") == ""
assert longest_palindrome("abc") in {"a", "b", "c"}
rng = random.Random(63)
for _ in range(40):
    s = "".join(rng.choice("abc") for _ in range(rng.randint(0, 10)))
    got = longest_palindrome(s)
    brute = _brute_palindromes(s)
    assert len(got) == len(brute) and got == got[::-1] and got in s
print("all checks passed")
```
