# 63. Longest Palindromic Substring

**Difficulty:** Medium. **Target:** O(n²) time. **Time box:** 25 minutes.

Return the longest palindromic substring of `s`. If several share that length, any one is fine.

DNA motif and text tools look for the same mirrored spans.

 Substrings are contiguous; a subsequence palindrome is a different problem.

**Constraints.** `s` length at most 20 in the random checks.


## Examples

- `"babad"` → `"bab"` or `"aba"`.
- `"cbbd"` → `"bb"`.
- `"a"` → `"a"`.
- `"ac"` → `"a"` or `"c"`.
