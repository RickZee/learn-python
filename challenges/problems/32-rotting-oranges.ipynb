# 32. Rotting Oranges

**Difficulty:** Medium. **Target:** O(rows·cols) time and space. **Time box:** 30 minutes.

Grid cells are empty (`0`), fresh orange (`1`), or rotten (`2`). Each minute every rotten orange simultaneously rots its four adjacent fresh oranges. Return the minutes until no fresh remain, or `-1` if some fresh orange never rots.

Infection and multi-source flood fills share this shape.

**Constraints.** The grid is at most 10 by 10.


## Examples

- `[[2,1,1],[1,1,0],[0,1,1]]` → `4`.
- `[[2,1,1],[0,1,1],[1,0,1]]` → `-1`.
- `[[0,2]]` → `0`.
- `[[1]]` → `-1`.
