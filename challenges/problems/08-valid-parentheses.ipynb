# 08. Valid Parentheses

**Difficulty:** Easy. **Target:** O(n) time, O(n) space. **Time box:** 20.

You are given a string `s` containing only `"()[]{}"`. Return whether the brackets are valid: every opener has a matching closer of the same type, and brackets nest correctly.

Parsers and editors check nesting the same way.

**Constraints.** `s` has 0 to 10_000 bracket characters.


## Examples

- `"()"` → `True`.
- `"()[]{}"` → `True`.
- `"(]"` → `False`.
- `"([)]"` → `False`.
- `""` → `True`.
