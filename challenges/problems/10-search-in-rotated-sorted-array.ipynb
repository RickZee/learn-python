# 10. Search in Rotated Sorted Array

**Difficulty:** Medium. **Target:** O(log n) time. **Time box:** 30.

You are given a list of distinct integers that was sorted ascending and then rotated at an unknown pivot. Return the index of `target`, or `-1` if missing. Still aim for logarithmic time.

Rotated search shows up when a ring buffer or versioned log is stored as one array.

**Constraints.** `nums` has 0 to 5000 distinct integers.


## Examples

- `[4,5,6,7,0,1,2]`, `0` → `4`.
- `[4,5,6,7,0,1,2]`, `3` → `-1`.
- `[1]`, `0` → `-1`.
- `[1]`, `1` → `0`.
