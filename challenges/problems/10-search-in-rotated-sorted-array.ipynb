```python
import random
import time


def search_rotated(nums: list[int], target: int) -> int:
    raise NotImplementedError

assert search_rotated([4, 5, 6, 7, 0, 1, 2], 0) == 4
assert search_rotated([4, 5, 6, 7, 0, 1, 2], 3) == -1
assert search_rotated([1], 0) == -1
assert search_rotated([1], 1) == 0
assert search_rotated([1, 3], 3) == 1
assert search_rotated([3, 1], 1) == 1
assert search_rotated([5, 1, 3], 5) == 0
assert search_rotated([4, 5, 6, 7, 8, 1, 2, 3], 8) == 4

def brute(nums, target):
    try:
        return nums.index(target)
    except ValueError:
        return -1

rng = random.Random(10)
for _ in range(40):
    base = sorted({rng.randint(0, 40) for _ in range(rng.randint(1, 15))})
    k = rng.randint(0, len(base) - 1)
    nums = base[k:] + base[:k]
    target = rng.randint(-1, 41)
    assert search_rotated(nums, target) == brute(nums, target)

big = list(range(50_000, 100_000)) + list(range(0, 50_000))
start = time.perf_counter()
assert search_rotated(big, 1234) == big.index(1234)
elapsed = time.perf_counter() - start
assert elapsed < 1.0
print(f"all checks passed ({elapsed * 1000:.0f} ms)")
```
