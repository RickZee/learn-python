## Examples

- `[1,2,3,4]` → `[24,12,8,6]`.
- `[-1,1,0,-3,3]` → `[0,0,9,0,0]` because of the single zero.
- `[2,3]` → `[3,2]`.

```python
import random
import time


def product_except_self(nums: list[int]) -> list[int]:
    raise NotImplementedError


def brute(nums):
    n = len(nums)
    out = []
    for i in range(n):
        p = 1
        for j in range(n):
            if i != j:
                p *= nums[j]
        out.append(p)
    return out


assert product_except_self([1, 2, 3, 4]) == [24, 12, 8, 6]
assert product_except_self([-1, 1, 0, -3, 3]) == [0, 0, 9, 0, 0]
assert product_except_self([2, 3]) == [3, 2]
assert product_except_self([0, 0]) == [0, 0]
assert product_except_self([9]) == [1]

rng = random.Random(5)
for _ in range(40):
    nums = [rng.randint(-3, 3) for _ in range(rng.randint(1, 8))]
    assert product_except_self(nums) == brute(nums)

big = [1, 2] * 10_000
start = time.perf_counter()
got = product_except_self(big)
assert len(got) == len(big)
elapsed = time.perf_counter() - start
assert elapsed < 1.0
print(f"all checks passed ({elapsed * 1000:.0f} ms)")
```
