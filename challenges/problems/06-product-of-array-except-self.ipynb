# 06. Product of Array Except Self

**Difficulty:** Medium. **Target:** O(n) time, O(1) extra space besides the answer. **Time box:** 25.

You are given a list of integers `nums`. Return a list `answer` where `answer[i]` is the product of every value in `nums` except `nums[i]`. Do not use division. The result fits in a normal Python int.

Feature scaling and prefix tricks show up the same way in data pipelines.

**Constraints.** `nums` has 0 to 10_000 integers.


## Examples

- `[1,2,3,4]` → `[24,12,8,6]`.
- `[-1,1,0,-3,3]` → `[0,0,9,0,0]`.
- `[2]` → `[1]` (empty product).
- `[]` → `[]`.
