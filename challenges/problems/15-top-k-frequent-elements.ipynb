# 15. Top K Frequent Elements

**Difficulty:** Medium. **Target:** O(n log k) time. **Time box:** 25 minutes.

You are given a list of integers `nums` and an integer `k`. Return the `k` values that appear most often. Order does not matter. It is guaranteed that the answer is unique.

Ranking popular items is a classic count-then-select problem.

**Constraints.** `nums` has 0 to 10_000 integers. `k` is between 0 and the number of distinct values.


## Examples

- `[1,1,1,2,2,3]`, `2` → `[1,2]` (any order).
- `[1]`, `1` → `[1]`.
- `[]`, `0` → `[]`.
- `[4,4,4,5,5,6]`, `1` → `[4]`.
