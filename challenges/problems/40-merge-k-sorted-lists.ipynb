# 40. Merge K Sorted Lists

**Difficulty:** Hard. **Target:** O(n log k) time, O(k) extra space. **Time box:** 35 minutes.

Merge `k` sorted linked lists into one sorted list. Lists arrive as arrays of values; your stub works on `ListNode` chains.

External sort and log merges face the same k-way merge.

**Constraints.** Up to 5 lists, each with at most 6 values, in the random checks.


## Examples

- `[[1,4,5],[1,3,4],[2,6]]` → `[1,1,2,3,4,4,5,6]`.
- `[]` → `[]`.
- `[[]]` → `[]`.
- `[[1],[0]]` → `[0,1]`.

````python
class ListNode:
    def __init__(self, val: int = 0, next: "ListNode | None" = None) -> None:
        self.val = val
        self.next = next


def nodes(values: list[int]) -> ListNode | None:
    dummy = ListNode()
    tail = dummy
    for value in values:
        tail.next = ListNode(value)
        tail = tail.next
    return dummy.next


def values(head: ListNode | None) -> list[int]:
    out: list[int] = []
    while head is not None:
        out.append(head.val)
        head = head.next
    return out


def merge_k_lists(lists: list[ListNode | None]) -> ListNode | None:
    raise NotImplementedError


from pylearn.challenges import cross_check


def merge_lists(rows: list[list[int]]) -> list[int]:
    return values(merge_k_lists([nodes(row) for row in rows]))


assert merge_lists([[1, 4, 5], [1, 3, 4], [2, 6]]) == [1, 1, 2, 3, 4, 4, 5, 6]
assert merge_lists([]) == []
assert merge_lists([[]]) == []
assert merge_lists([[1], [0]]) == [0, 1]
assert merge_lists([[1, 2, 3]]) == [1, 2, 3]
assert merge_lists([[], [1], []]) == [1]
assert merge_lists([[5, 6], [1, 2], [3]]) == [1, 2, 3, 5, 6]
assert merge_lists([[1, 1], [1]]) == [1, 1, 1]
cross_check("40-merge-k-sorted-lists", merge_lists)
print("all checks passed")
````
