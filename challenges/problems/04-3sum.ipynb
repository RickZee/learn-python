# 04. 3Sum

**Difficulty:** Medium. **Target:** O(n²) time after sort. **Time box:** 35.

You are given a list of integers `nums`. Return all unique triplets `[a, b, c]` such that `a + b + c == 0`. Triplets must use distinct indexes. Order of triplets does not matter; values inside a triplet should be non-decreasing.

Finding zero-sum combinations shows up in geometry and transaction matching.

**Constraints.** Up to 1000 integers. Values fit in a normal Python int.


## Examples

- `[-1,0,1,2,-1,-4]` → `[[-1,-1,2],[-1,0,1]]`.
- `[0,1,1]` → `[]`.
- `[0,0,0]` → `[[0,0,0]]`.
- `[]` → `[]`.
