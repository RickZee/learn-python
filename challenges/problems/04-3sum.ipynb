## Examples

- `[0,1,1]` → `[]` — no three numbers sum to zero.
- `[0,0,0]` → `[[0,0,0]]`.

```python
import random


def three_sum(nums: list[int]) -> list[list[int]]:
    raise NotImplementedError


def _norm(triplets):
    return sorted(sorted(t) for t in triplets)

assert _norm(three_sum([-1, 0, 1, 2, -1, -4])) == _norm([[-1, -1, 2], [-1, 0, 1]])
assert _norm(three_sum([0, 1, 1])) == []
assert _norm(three_sum([0, 0, 0])) == [[0, 0, 0]]
assert _norm(three_sum([])) == []
assert _norm(three_sum([1, 2])) == []
assert _norm(three_sum([-2, 0, 1, 1, 2])) == _norm([[-2, 0, 2], [-2, 1, 1]])
assert _norm(three_sum([1, 1, -2])) == [[-2, 1, 1]]
assert _norm(three_sum([3, -2, 1, 0])) == []

def brute(nums):
    n = len(nums)
    found = set()
    for i in range(n):
        for j in range(i + 1, n):
            for k in range(j + 1, n):
                if nums[i] + nums[j] + nums[k] == 0:
                    found.add(tuple(sorted((nums[i], nums[j], nums[k]))))
    return [list(t) for t in found]

rng = random.Random(3)
for _ in range(25):
    nums = [rng.randint(-8, 8) for _ in range(rng.randint(0, 10))]
    assert _norm(three_sum(nums)) == _norm(brute(nums))
print("all checks passed")
```
