# 55. Permutations

**Difficulty:** Medium. **Target:** O(n · n!) time. **Time box:** 20 minutes.

Return every permutation of the distinct integers in `nums`. Order of the permutations in the output does not matter.

Backtracking over swaps generates the symmetric group.

 Each output row is a full rearrangement; do not drop or duplicate any input value.

**Constraints.** `nums` has distinct values; length at most 5 in checks.


## Examples

- `[1,2,3]` → all 6 permutations.
- `[0,1]` → `[[0,1],[1,0]]`.
- `[1]` → `[[1]]`.
