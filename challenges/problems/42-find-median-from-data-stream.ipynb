# 42. Find Median from Data Stream

**Difficulty:** Hard. **Target:** O(log n) add, O(1) read. **Time box:** 35 minutes.

Build a structure that accepts numbers with `add_num` and returns the current median with `find_median`. For an even count, return the average of the two middle values as a float.

Live dashboards keep a running median of a stream.

**Constraints.** Values are small integers. Random op sequences stay short.


## Examples

- add 1, add 2, find_median → `1.5`; add 3, find_median → `2.0`.
- add 5, find_median → `5.0`.


## Before you code

1. How do two heaps split the lower and upper halves?
2. What balance invariant keeps the median at the heap tops?
3. How do you average when both halves have the same size?

<details><summary>Hint 1: the pattern</summary>

Design: max-heap for the low half, min-heap for the high half.

<details><summary>Hint 2: the approach</summary>
