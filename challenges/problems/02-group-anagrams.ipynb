## Examples

- `["eat","tea","tan"]` → groups `["eat","tea"]` and `["tan"]` because eat/tea share letters.
- `["a"]` → `[["a"]]` — one word is one group.
- `[""]` → `[[""]]` — empty strings are anagrams of each other.
