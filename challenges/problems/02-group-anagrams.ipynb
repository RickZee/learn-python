# 02. Group Anagrams

**Difficulty:** Medium. **Target:** O(n·k log k) time, O(n·k) space. **Time box:** 25.

You are given a list of strings `words`. Group the ones that are anagrams of each other. Two words are anagrams when they use the same letters the same number of times. Return the groups in any order.

Search and spell-check features often bucket words this way before scoring them.

**Constraints.** Up to 1000 words. Each word has 0 to 20 lowercase English letters.


## Examples

- `["eat","tea","tan","ate","nat","bat"]` → three groups: the eat/tea/ate cluster, tan/nat, and bat.
- `[""]` → `[[""]]`.
- `["a"]` → `[["a"]]`.
- `[]` → `[]`.
