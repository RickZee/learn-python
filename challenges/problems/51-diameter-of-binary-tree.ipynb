# 51. Diameter of Binary Tree

**Difficulty:** Easy. **Target:** O(n) time. **Time box:** 20 minutes.

The diameter is the number of *edges* on the longest path between any two nodes. The tree is given as a list in heap layout: the children of index `i` sit at `2*i+1` and `2*i+2`, with `None` for missing nodes.

Network hops across a hierarchy ask for the same longest path.

**Constraints.** Up to 15 list slots in the random checks.


## Examples

- `[1,2,3,4,5]` → `3` (path 4–2–1–3 or 5–2–1–3).
- `[1,2]` → `1`.
- `[1]` → `0`.
- `[]` → `0`.
