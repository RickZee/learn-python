```python
import random


def diameter_of_values(*args, **kwargs):
    raise NotImplementedError


def _brute_diameter(values):
    n = len(values)
    best = 0

    def depth(i):
        nonlocal best
        if i >= n or values[i] is None:
            return 0
        left = depth(2 * i + 1)
        right = depth(2 * i + 2)
        best = max(best, left + right)
        return 1 + max(left, right)

    depth(0)
    return best


assert diameter_of_values([1, 2, 3, 4, 5]) == 3
assert diameter_of_values([1, 2]) == 1
assert diameter_of_values([1]) == 0
assert diameter_of_values([1, None, 2]) == 1
assert diameter_of_values([1, 2, 3]) == 2
assert diameter_of_values([1, 2, None, 3]) == 2
assert diameter_of_values([1, 2, 3, 4, None, None, 5]) == 4
assert diameter_of_values([1, None, None]) == 0
rng = random.Random(51)
for _ in range(40):
    size = rng.randint(1, 15)
    values = []
    for i in range(size):
        if i == 0 or (values[(i - 1) // 2] is not None and rng.random() < 0.7):
            values.append(rng.randint(0, 9))
        else:
            values.append(None)
    assert diameter_of_values(values) == _brute_diameter(values)
print("all checks passed")
```
