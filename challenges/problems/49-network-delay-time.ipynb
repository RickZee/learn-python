```python
import random


def network_delay(*args, **kwargs):
    raise NotImplementedError


def _brute_network_delay(times, n, k):
    dist = [[10**9] * (n + 1) for _ in range(n + 1)]
    for i in range(1, n + 1):
        dist[i][i] = 0
    for u, v, w in times:
        dist[u][v] = min(dist[u][v], w)
    for mid in range(1, n + 1):
        for i in range(1, n + 1):
            for j in range(1, n + 1):
                dist[i][j] = min(dist[i][j], dist[i][mid] + dist[mid][j])
    worst = max(dist[k][i] for i in range(1, n + 1))
    return -1 if worst >= 10**9 else worst


assert network_delay([[2, 1, 1], [2, 3, 1], [3, 4, 1]], 4, 2) == 2
assert network_delay([[1, 2, 1]], 2, 1) == 1
assert network_delay([[1, 2, 1]], 2, 2) == -1
assert network_delay([[1, 2, 1], [2, 3, 2], [1, 3, 4]], 3, 1) == 3
assert network_delay([[1, 2, 1], [2, 1, 3]], 2, 2) == 3
assert network_delay([[1, 2, 1], [2, 3, 1], [3, 1, 1]], 3, 1) == 2
assert network_delay([[1, 2, 1]], 2, 1) == 1
assert network_delay([[1, 2, 1], [1, 3, 1]], 3, 1) == 1
rng = random.Random(49)
for _ in range(30):
    n = rng.randint(1, 5)
    times = []
    for u in range(1, n + 1):
        for v in range(1, n + 1):
            if u != v and rng.random() < 0.35:
                times.append([u, v, rng.randint(1, 5)])
    k = rng.randint(1, n)
    assert network_delay(times, n, k) == _brute_network_delay(times, n, k)
print("all checks passed")
```
