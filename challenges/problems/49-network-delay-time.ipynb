# 49. Network Delay Time

**Difficulty:** Medium. **Target:** O((V+E) log V). **Time box:** 30 minutes.

`times` lists directed edges `[u, v, w]` meaning a signal from node `u` to `v` takes `w` time. Nodes are labeled `1..n`. A signal starts at node `k`. Return how long until every node receives it, or `-1` if some node is unreachable.

Broadcast latency on a weighted digraph is single-source shortest paths.

**Constraints.** `n` is at most 8 in the random checks.


## Examples

- `[[2,1,1],[2,3,1],[3,4,1]]`, `n = 4`, `k = 2` → `2`.
- `[[1,2,1]]`, `n = 2`, `k = 1` → `1`.
- `[[1,2,1]]`, `n = 2`, `k = 2` → `-1`.
