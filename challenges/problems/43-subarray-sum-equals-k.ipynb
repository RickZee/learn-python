**Difficulty:** Medium. **Target:** O(n) time, O(n) space. **Time box:** 25 minutes.

Count how many contiguous subarrays sum to `k`. The array may contain negatives, so a simple positive-only sliding window is not enough.

Prefix-sum maps turn range-sum queries into lookups.

**Constraints.** `nums` has up to 12 integers in the random checks.


## Examples

- `[1,1,1]`, `k = 2` → `2`.
- `[1,2,3]`, `k = 3` → `2`.
- `[1]`, `k = 0` → `0`.
- `[-1,-1,1]`, `k = 0` → `1`.
