```python
import random


def subarray_sum(*args, **kwargs):
    raise NotImplementedError


def _brute_subarray_sum(nums, k):
    total = 0
    for left in range(len(nums)):
        running = 0
        for right in range(left, len(nums)):
            running += nums[right]
            if running == k:
                total += 1
    return total


assert subarray_sum([1, 1, 1], 2) == 2
assert subarray_sum([1, 2, 3], 3) == 2
assert subarray_sum([1], 0) == 0
assert subarray_sum([-1, -1, 1], 0) == 1
assert subarray_sum([1, 2, 1, 2, 1], 3) == 4
assert subarray_sum([], 1) == 0
assert subarray_sum([0, 0, 0], 0) == 6
assert subarray_sum([3, 4, 7, 2, -3, 1, 4, 2], 7) == 4
rng = random.Random(43)
for _ in range(40):
    nums = [rng.randint(-5, 5) for _ in range(rng.randint(0, 10))]
    k = rng.randint(-8, 8)
    assert subarray_sum(nums, k) == _brute_subarray_sum(nums, k)
print("all checks passed")
```
