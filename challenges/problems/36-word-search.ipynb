# 36. Word Search

**Difficulty:** Medium. **Target:** O(rows·cols·4^L) time. **Time box:** 30 minutes.

Given a board of letters and a word, return whether the word appears as a path of four-adjacent cells (no diagonals). Each cell may be used at most once in the path.

Grid word puzzles are the same backtracking walk.

**Constraints.** Boards are at most 4 by 4 in the random checks. Words are short.


## Examples

- Board `A B C E / S F C S / A D E E`, word `ABCCED` → `True`.
- Same board, `SEE` → `True`.
- Same board, `ABCB` → `False`.
- Board `A`, word `A` → `True`.
