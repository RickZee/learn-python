```python
import random


def length_of_lis(*args, **kwargs):
    raise NotImplementedError


def _brute_lis(nums):
    n = len(nums)
    if n == 0:
        return 0
    best = [1] * n
    for i in range(n):
        for j in range(i):
            if nums[j] < nums[i]:
                best[i] = max(best[i], best[j] + 1)
    return max(best)


assert length_of_lis([10, 9, 2, 5, 3, 7, 101, 18]) == 4
assert length_of_lis([0, 1, 0, 3, 2, 3]) == 4
assert length_of_lis([7, 7, 7, 7]) == 1
assert length_of_lis([1]) == 1
assert length_of_lis([1, 2, 3]) == 3
assert length_of_lis([3, 2, 1]) == 1
assert length_of_lis([1, 3, 2, 4]) == 3
assert length_of_lis([]) == 0
rng = random.Random(54)
for _ in range(40):
    nums = [rng.randint(-10, 10) for _ in range(rng.randint(0, 12))]
    assert length_of_lis(nums) == _brute_lis(nums)
print("all checks passed")
```
