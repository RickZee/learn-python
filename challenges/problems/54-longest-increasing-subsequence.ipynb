**Difficulty:** Medium. **Target:** O(n log n). **Time box:** 30 minutes.

Return the length of the longest *strictly* increasing subsequence of `nums`. The subsequence need not be contiguous.

Patience sorting and LIS show up in version timelines and ranking.

 Contiguous runs are subarrays, not subsequences; order must stay strictly rising.

**Constraints.** `nums` has up to 12 integers in the random checks.


## Examples

- `[10,9,2,5,3,7,101,18]` → `4`.
- `[0,1,0,3,2,3]` → `4`.
- `[7,7,7,7]` → `1`.
- `[]` → `0`.
