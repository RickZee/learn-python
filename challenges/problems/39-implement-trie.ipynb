# 39. Implement Trie

**Difficulty:** Medium. **Target:** O(L) per call. **Time box:** 30 minutes.

Implement a trie (prefix tree) with `insert`, `search`, and `starts_with`. `search` is true only for a full inserted word; `starts_with` is true for any prefix of an inserted word.

Autocomplete and spell-check indexes are tries.

**Constraints.** Words use lowercase letters. Random checks use short words.


## Examples

- insert `apple`; search `apple` → `True`; search `app` → `False`; starts_with `app` → `True`; insert `app`; search `app` → `True`.

2. How does search differ from starts_with at the end of the walk?
3. Why is a flat set of words a weak starts_with answer?

Design: child map per node plus an end-of-word mark.

<details><summary>Hint 2: the approach</summary>

Walk character by character. insert creates missing children. search requires the end mark.
</details>
