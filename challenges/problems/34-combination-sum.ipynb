# 34. Combination Sum

**Difficulty:** Medium. **Target:** exponential time, O(target) stack. **Time box:** 30 minutes.

Given distinct positive candidates and a target, list every combination that sums to the target. You may reuse a candidate. Order inside a combination does not create duplicates — `[2,2,3]` counts once.

Making change with unlimited coins, listing every way, is the same search.

**Constraints.** Candidates are distinct positives. Target is at most 30 in checks.


## Examples

- `[2,3,6,7]`, `7` → `[[2,2,3],[7]]` (any order of combinations).
- `[2,3,5]`, `8` → `[[2,2,2,2],[2,3,3],[3,5]]`.
- `[2]`, `1` → `[]`.
- `[1]`, `1` → `[[1]]`.
