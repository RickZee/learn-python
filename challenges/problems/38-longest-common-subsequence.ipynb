**Difficulty:** Medium. **Target:** O(m·n) time, O(n) extra space. **Time box:** 30 minutes.

Diff tools and DNA comparisons lean on this length.

**Constraints.** Each string has 0 to 12 lowercase letters in the random checks.


## Examples

- `"abcde"`, `"ace"` → `3`.
- `"abc"`, `"abc"` → `3`.
- `"abc"`, `"def"` → `0`.
- `""`, `"a"` → `0`.
