# 37. Unique Paths

**Difficulty:** Medium. **Target:** O(m·n) time, O(n) extra space. **Time box:** 25 minutes.

A robot sits at the top-left of an `m` by `n` grid and may move only right or down. Return how many distinct paths reach the bottom-right cell.

Grid path counts show up in routing and in counting lattice walks.

**Constraints.** `m` and `n` are between 1 and 10 in the random checks.


## Examples

- `m = 3`, `n = 7` → `28`.
- `m = 3`, `n = 2` → `3`.
- `m = 1`, `n = 1` → `1`.
- `m = 2`, `n = 3` → `3`.
