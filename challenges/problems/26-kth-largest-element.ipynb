# 26. Kth Largest Element

**Difficulty:** Medium. **Target:** O(n log k) time, O(k) extra space. **Time box:** 25 minutes.

Return the kth largest value in an unsorted array. Duplicates count as separate occurrences: in `[3,2,3,1,2,4,5,5,6]` with `k = 4` the answer is `4`.

Leaderboards and order statistics need the boundary of the top k, not a full sort.

**Constraints.** `nums` has 1 to 100 integers. `k` is between 1 and `len(nums)`.


## Examples

- `[3,2,1,5,6,4]`, `k = 2` → `5`.
- `[3,2,3,1,2,4,5,5,6]`, `k = 4` → `4`.
- `[1]`, `k = 1` → `1`.
- `[2,1]`, `k = 2` → `1`.
