```python
import random
import time


def max_area(height: list[int]) -> int:
    raise NotImplementedError


def brute(height):
    best = 0
    for i in range(len(height)):
        for j in range(i + 1, len(height)):
            best = max(best, min(height[i], height[j]) * (j - i))
    return best


assert max_area([1, 8, 6, 2, 5, 4, 8, 3, 7]) == 49
assert max_area([1, 1]) == 1
assert max_area([4, 3, 2, 1, 4]) == 16
assert max_area([1, 2, 1]) == 2
assert max_area([2, 3, 4, 5, 18, 17, 6]) == 17

rng = random.Random(4)
for _ in range(40):
    height = [rng.randint(0, 20) for _ in range(rng.randint(2, 12))]
    assert max_area(height) == brute(height)

big = [1 + (i % 50) for i in range(30_000)]
start = time.perf_counter()
got = max_area(big)
elapsed = time.perf_counter() - start
assert got > 0
assert elapsed < 1.0, f"{elapsed:.2f}s"
print(f"all checks passed ({elapsed * 1000:.0f} ms)")
```
