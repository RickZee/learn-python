**Difficulty:** Medium. **Target:** O(n) time, O(1) extra space. **Time box:** 25.

You are given a list `height` where `height[i]` is the height of a vertical line at x=`i`. Choose two lines so the container they form with the x-axis holds the most water. Area is `min(height[l], height[r]) * (r - l)`. Return that maximum area.

Geometry warm-ups and capacity problems use the same two-pointer shrink.

**Constraints.** `height` has 0 to 10_000 non-negative integers.


## Examples

- `[1,8,6,2,5,4,8,3,7]` → `49`.
- `[1,1]` → `1`.
- `[1]` → `0`.
