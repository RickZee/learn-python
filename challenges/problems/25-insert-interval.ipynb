# 25. Insert Interval

**Difficulty:** Medium. **Target:** O(n) time, O(n) extra space. **Time box:** 25 minutes.

You are given a list of non-overlapping intervals sorted by start, and one new interval `new`. Insert `new` and merge whatever overlaps so the result stays sorted and non-overlapping.

Calendars do this when you book a block that touches existing events.

**Constraints.** Each interval is `[start, end]` with `start <= end`. The input list has 0 to 100 intervals.


## Examples

- `[[1,3],[6,9]]`, `[2,5]` → `[[1,5],[6,9]]` (merge the first).
- `[[1,2],[3,5],[6,7],[8,10],[12,16]]`, `[4,8]` → `[[1,2],[3,10],[12,16]]`.
- `[]`, `[5,7]` → `[[5,7]]`.
- `[[1,5]]`, `[2,3]` → `[[1,5]]` (new sits inside).
