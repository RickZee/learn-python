## Examples

- `[[1,3],[2,6],[8,10],[15,18]]` → `[[1,6],[8,10],[15,18]]` because 1–3 and 2–6 overlap.
- `[[1,4],[4,5]]` → `[[1,5]]` because they touch.
- `[]` → `[]`.
