# 07. Merge Intervals

**Difficulty:** Medium. **Target:** O(n log n) time. **Time box:** 25.

You are given a list of closed intervals `[start, end]`. Merge every overlapping or touching pair and return the list of disjoint intervals sorted by start.

Calendar free/busy and range compression both need this merge.

**Constraints.** Up to 1000 intervals. Ends are integers with start ≤ end.


## Examples

- `[[1,3],[2,6],[8,10],[15,18]]` → `[[1,6],[8,10],[15,18]]`.
- `[[1,4],[4,5]]` → `[[1,5]]` (touching merges).
- `[]` → `[]`.
- `[[1,4]]` → `[[1,4]]`.
