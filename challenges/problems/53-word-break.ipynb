```python
import random


def word_break(*args, **kwargs):
    raise NotImplementedError


def _brute_word_break(s, words):
    wordset = set(words)

    def ok(i):
        if i == len(s):
            return True
        return any(s[i:j] in wordset and ok(j) for j in range(i + 1, len(s) + 1))

    return ok(0)


assert word_break("leetcode", ["leet", "code"]) is True
assert word_break("applepenapple", ["apple", "pen"]) is True
assert word_break("catsandog", ["cats", "dog", "sand", "and", "cat"]) is False
assert word_break("a", ["a"]) is True
assert word_break("a", ["b"]) is False
assert word_break("bb", ["a", "b", "bbb"]) is True
assert word_break("", ["a"]) is True
assert word_break("abcd", ["a", "abc", "b", "cd"]) is True
rng = random.Random(53)
alphabet = "abcd"
for _ in range(40):
    words = list(
        {
            "".join(rng.choice(alphabet) for _ in range(rng.randint(1, 3)))
            for _ in range(rng.randint(1, 5))
        }
    )
    if rng.random() < 0.7:
        s = "".join(rng.choice(words) for _ in range(rng.randint(0, 4)))
    else:
        s = "".join(rng.choice(alphabet) for _ in range(rng.randint(0, 8)))
    assert word_break(s, words) is _brute_word_break(s, words)
print("all checks passed")
```
