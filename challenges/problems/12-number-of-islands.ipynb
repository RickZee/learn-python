# 12. Number of Islands

**Difficulty:** Medium. **Target:** O(rows·cols) time. **Time box:** 30.

You are given a 2D grid of `"1"` (land) and `"0"` (water). An island is a maximal 4-directionally connected group of land cells. Return how many islands the grid contains.

Connected components on a grid are the same idea as flood fill in image tools.

**Constraints.** Grid has 0 to 300 rows and columns.


## Examples

- `[["1","1","0"],["1","1","0"],["0","0","1"]]` → `2`.
- `[["1"]]` → `1`.
- `[["0"]]` → `0`.
- `[]` → `0`.
