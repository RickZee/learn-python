- `[["0","0"],["0","0"]]` → `0`.

```python
import copy
import random


def num_islands(grid: list[list[str]]) -> int:
    raise NotImplementedError

assert num_islands([["1","1","0"],["1","0","0"],["0","0","1"]]) == 2
assert num_islands([["1"]]) == 1
assert num_islands([["0","0"],["0","0"]]) == 0
assert num_islands([]) == 0
assert num_islands([["1","1","1"],["0","1","0"],["1","1","1"]]) == 1
assert num_islands([["1","0","1","0","1"]]) == 3
assert num_islands([["1"],["0"],["1"]]) == 2
assert num_islands([["1","1"],["1","1"]]) == 1

def brute(grid):
    g = copy.deepcopy(grid)
    if not g or not g[0]:
        return 0
    rows, cols = len(g), len(g[0])
    def sink(r, c):
        if r < 0 or c < 0 or r >= rows or c >= cols or g[r][c] != "1":
            return
        g[r][c] = "0"
        for dr, dc in ((1,0),(-1,0),(0,1),(0,-1)):
            sink(r+dr, c+dc)
    count = 0
    for r in range(rows):
        for c in range(cols):
            if g[r][c] == "1":
                count += 1
                sink(r, c)
    return count

rng = random.Random(12)
for _ in range(30):
    rows, cols = rng.randint(1, 6), rng.randint(1, 6)
    grid = [[rng.choice(["0", "1"]) for _ in range(cols)] for _ in range(rows)]
    assert num_islands(copy.deepcopy(grid)) == brute(grid)
print("all checks passed")
```
