```python
import random


def num_decodings(*args, **kwargs):
    raise NotImplementedError


def _brute_decode(s):
    memo = {}

    def go(i):
        if i == len(s):
            return 1
        if s[i] == "0":
            return 0
        if i in memo:
            return memo[i]
        ways = go(i + 1)
        if i + 1 < len(s) and 10 <= int(s[i : i + 2]) <= 26:
            ways += go(i + 2)
        memo[i] = ways
        return ways

    return 0 if not s else go(0)


assert num_decodings("12") == 2
assert num_decodings("226") == 3
assert num_decodings("06") == 0
assert num_decodings("10") == 1
assert num_decodings("27") == 1
assert num_decodings("1") == 1
assert num_decodings("2101") == 1
assert num_decodings("11106") == 2
rng = random.Random(58)
for _ in range(40):
    s = "".join(str(rng.randint(0, 9)) for _ in range(rng.randint(1, 8)))
    assert num_decodings(s) == _brute_decode(s)
print("all checks passed")
```
