# 58. Decode Ways

**Difficulty:** Medium. **Target:** O(n) time, O(1) space. **Time box:** 25 minutes.

A message encoding maps `A→1` … `Z→26`. Given a digit string `s`, return how many ways it can be decoded. Leading zeros are invalid (`0` alone or `06`).

Counting parses of a digit stream is classic rolling DP.

**Constraints.** `s` length at most 10 in the random checks; digits only.


## Examples

- `"12"` → `2` (`AB` or `L`).
- `"226"` → `3`.
- `"06"` → `0`.
- `"10"` → `1`.
