```python
import random


def remove_nth_from_end_list(*args, **kwargs):
    raise NotImplementedError


def _brute_remove(values, n):
    idx = len(values) - n
    return values[:idx] + values[idx + 1 :]


assert remove_nth_from_end_list([1, 2, 3, 4, 5], 2) == [1, 2, 3, 5]
assert remove_nth_from_end_list([1], 1) == []
assert remove_nth_from_end_list([1, 2], 1) == [1]
assert remove_nth_from_end_list([1, 2], 2) == [2]
assert remove_nth_from_end_list([1, 2, 3], 3) == [2, 3]
assert remove_nth_from_end_list([1, 2, 3], 1) == [1, 2]
assert remove_nth_from_end_list([1, 2, 3, 4], 4) == [2, 3, 4]
assert remove_nth_from_end_list([0, 1], 1) == [0]
rng = random.Random(56)
for _ in range(40):
    values = [rng.randint(0, 9) for _ in range(rng.randint(1, 12))]
    n = rng.randint(1, len(values))
    assert remove_nth_from_end_list(values, n) == _brute_remove(values, n)
print("all checks passed")
```
