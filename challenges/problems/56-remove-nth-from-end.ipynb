# 56. Remove Nth Node From End of List

**Difficulty:** Medium. **Target:** O(L) time, O(1) pointers. **Time box:** 20 minutes.

Given a list of values standing in for a linked list, remove the n-th node from the end and return the remaining values. n is 1-based from the end.

Two pointers with a fixed gap delete in one pass on a real list.

**Constraints.** The list has at least n nodes. Lengths stay under 12 in checks.


## Examples

- `[1,2,3,4,5]`, `n = 2` → `[1,2,3,5]`.
- `[1]`, `n = 1` → `[]`.
- `[1,2]`, `n = 1` → `[1]`.
- `[1,2]`, `n = 2` → `[2]`.
