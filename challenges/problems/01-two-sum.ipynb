# 01. Two Sum

**Difficulty:** Easy. **Target:** O(n) time, O(n) extra space. **Time box:** 20 minutes.

You are given a list of integers `nums` and an integer `target`. Return the indexes of two different elements whose values add up to `target`. Return them with the smaller index first. If no such pair exists, return an empty list. There is at most one valid pair.

Looking up a complement while you walk the list once is the usual interview warm-up for hash maps.

**Constraints.** `nums` has 0 to 10_000 integers. Each value and `target` fit in a normal Python `int`.


## Examples

- `[2, 7, 11, 15]`, `9` → `[0, 1]` because `2 + 7 = 9`.
- `[3, 2, 4]`, `6` → `[1, 2]` because `2 + 4 = 6` (not the first `3`).
- `[3, 3]`, `6` → `[0, 1]`: two different indexes, even when the values match.
- `[1]`, `2` → `[]`: a single element cannot form a pair.

3. If the same value appears twice, when do you write the second index into the dict?


## Hints

<details><summary>Hint 1: the pattern</summary>

</details>

<details><summary>Hint 2: the approach</summary>

Walk left to right. For value `v` at index `i`, look up `target - v` in a dict of earlier values to indexes. If it is there, return those two indexes (smaller first). Otherwise store `v → i` and continue. If the walk ends, return `[]`.
</details>
