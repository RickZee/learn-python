# 14. Merge Two Sorted Lists

**Difficulty:** Easy. **Target:** O(n+m) time, O(1) extra space. **Time box:** 20 minutes.

You are given the heads of two sorted singly linked lists. Merge them into one sorted list by splicing the existing nodes. Either input may be empty.

Merging sorted runs is the combine step of merge sort on lists.

**Constraints.** Each list has 0 to 50 nodes with non-decreasing values.


## Examples

- `[1,2,4]` and `[1,3,4]` → `[1,1,2,3,4,4]`.
- `[]` and `[]` → `[]`.
- `[]` and `[0]` → `[0]`.
- `[2]` and `[1]` → `[1,2]`.
