```python
import random


def can_jump(*args, **kwargs):
    raise NotImplementedError


def _brute_can_jump(nums):
    n = len(nums)
    reach = [False] * n
    reach[0] = True
    for i in range(n):
        if not reach[i]:
            continue
        for step in range(1, nums[i] + 1):
            if i + step < n:
                reach[i + step] = True
            else:
                return True
    return reach[-1] if n else True


assert can_jump([2, 3, 1, 1, 4]) is True
assert can_jump([3, 2, 1, 0, 4]) is False
assert can_jump([0]) is True
assert can_jump([1, 0]) is True
assert can_jump([2, 0, 0]) is True
assert can_jump([1, 0, 1]) is False
assert can_jump([5]) is True
assert can_jump([1, 1, 1, 0]) is True
rng = random.Random(47)
for _ in range(40):
    nums = [rng.randint(0, 4) for _ in range(rng.randint(1, 10))]
    assert can_jump(nums) is _brute_can_jump(nums)
print("all checks passed")
```
