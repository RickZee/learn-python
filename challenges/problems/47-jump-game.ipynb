# 47. Jump Game

**Difficulty:** Medium. **Target:** O(n) time, O(1) space. **Time box:** 20 minutes.

Each index holds a jump length. Starting at index 0, return whether you can reach the last index. You may land anywhere within the jump length.

Reachability along a line is a one-pass farthest-reach scan.

**Constraints.** `nums` has 1 to 15 non-negative integers in the checks.


## Examples

- `[2,3,1,1,4]` → `True`.
- `[3,2,1,0,4]` → `False`.
- `[0]` → `True`.
- `[2,0,0]` → `True`.
