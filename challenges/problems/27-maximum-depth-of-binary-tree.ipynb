# 27. Maximum Depth of Binary Tree

**Difficulty:** Easy. **Target:** O(n) time, O(n) stack space. **Time box:** 15 minutes.

Depth is the number of nodes on the longest root-to-leaf path. An empty tree has depth 0. Trees arrive as a level-order list with `None` for missing children.

Folder trees and org charts ask the same height question.

**Constraints.** The tree has 0 to 20 nodes. Values are integers.


## Examples

- `[3,9,20,None,None,15,7]` → `3`.
- `[1,None,2]` → `2`.
- `[]` → `0`.
- `[1]` → `1`.

````python
class TreeNode:
    def __init__(
        self,
        val: int = 0,
        left: "TreeNode | None" = None,
        right: "TreeNode | None" = None,
    ) -> None:
        self.val = val
        self.left = left
        self.right = right


def tree(values: list[int | None]) -> TreeNode | None:
    if not values or values[0] is None:
        return None
    root = TreeNode(values[0])
    queue: list[TreeNode] = [root]
    index = 1
    while queue and index < len(values):
        node = queue.pop(0)
        if index < len(values) and values[index] is not None:
            node.left = TreeNode(values[index])  # type: ignore[arg-type]
            queue.append(node.left)
        index += 1
        if index < len(values) and values[index] is not None:
            node.right = TreeNode(values[index])  # type: ignore[arg-type]
            queue.append(node.right)
        index += 1
    return root


def flatten(root: TreeNode | None) -> list[int | None]:
    if root is None:
        return []
    out: list[int | None] = []
    queue: list[TreeNode | None] = [root]
    while queue:
        node = queue.pop(0)
        if node is None:
            out.append(None)
            continue
        out.append(node.val)
        queue.append(node.left)
        queue.append(node.right)
    while out and out[-1] is None:
        out.pop()
    return out


def max_depth(root: TreeNode | None) -> int:
    raise NotImplementedError


from pylearn.challenges import cross_check

assert max_depth(tree([3, 9, 20, None, None, 15, 7])) == 3
assert max_depth(tree([1, None, 2])) == 2
assert max_depth(tree([])) == 0
assert max_depth(tree([1])) == 1
assert max_depth(tree([1, 2, 3, 4, 5])) == 3
assert max_depth(tree([1, 2, None, 3])) == 3
assert max_depth(tree([1, None, None])) == 1
assert max_depth(tree([2, 1, 3])) == 2
cross_check("27-maximum-depth-of-binary-tree", lambda vals: max_depth(tree(vals)))
print("all checks passed")
````
