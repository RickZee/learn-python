```python
import random


def min_meeting_rooms(*args, **kwargs):
    raise NotImplementedError


def _brute_rooms(intervals):
    if not intervals:
        return 0
    events = []
    for start, end in intervals:
        events.append((start, 1))
        events.append((end, -1))
    events.sort(key=lambda e: (e[0], e[1]))
    cur = best = 0
    for _, delta in events:
        cur += delta
        best = max(best, cur)
    return best


assert min_meeting_rooms([[0, 30], [5, 10], [15, 20]]) == 2
assert min_meeting_rooms([[7, 10], [2, 4]]) == 1
assert min_meeting_rooms([]) == 0
assert min_meeting_rooms([[1, 5]]) == 1
assert min_meeting_rooms([[1, 10], [2, 6], [3, 5]]) == 3
assert min_meeting_rooms([[1, 2], [2, 3]]) == 1
assert min_meeting_rooms([[1, 4], [2, 5], [3, 6]]) == 3
assert min_meeting_rooms([[0, 1], [1, 2], [2, 3]]) == 1
rng = random.Random(68)
for _ in range(40):
    intervals = []
    for __ in range(rng.randint(0, 8)):
        start = rng.randint(0, 20)
        end = start + rng.randint(1, 10)
        intervals.append([start, end])
    assert min_meeting_rooms(intervals) == _brute_rooms(intervals)
print("all checks passed")
```
