# 64. Task Scheduler

**Difficulty:** Medium. **Target:** O(T) time. **Time box:** 25 minutes.

You have CPU tasks labeled by letters and a cool-down `n`: two equal tasks must be at least `n` intervals apart. Idle slots fill the gaps. Return the minimum number of intervals to finish every task.

Batch jobs with anti-affinity cool-downs use the same math.

**Constraints.** Tasks are uppercase letters. Lengths stay under 20 in checks.


## Examples

- `["A","A","A","B","B","B"]`, `n = 2` → `8`.
- `["A","A","A","B","B","B"]`, `n = 0` → `6`.
- `["A","A","A","A","B","B","C","C"]`, `n = 1`? check formula.
- `["A"]`, `n = 2` → `1`.
