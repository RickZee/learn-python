```python
import random
from collections import Counter


def least_interval(*args, **kwargs):
    raise NotImplementedError


def _brute_least_interval(tasks, n):
    counts = list(Counter(tasks).values())
    peak = max(counts)
    peak_count = counts.count(peak)
    return max(len(tasks), (peak - 1) * (n + 1) + peak_count)


assert least_interval(["A", "A", "A", "B", "B", "B"], 2) == 8
assert least_interval(["A", "C", "A", "B", "D", "B"], 1) == 6
assert least_interval(["A", "A", "A", "B", "B", "B"], 0) == 6
assert least_interval(["A"], 1) == 1
assert least_interval(["A", "A"], 2) == 4
assert least_interval(["A", "B"], 0) == 2
assert least_interval(["A", "A", "A"], 1) == 5
assert least_interval(["A", "B", "C"], 2) == 3
rng = random.Random(64)
for _ in range(40):
    tasks = [rng.choice("ABCD") for _ in range(rng.randint(1, 12))]
    n = rng.randint(0, 5)
    assert least_interval(tasks, n) == _brute_least_interval(tasks, n)
print("all checks passed")
```
