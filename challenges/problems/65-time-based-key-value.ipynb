## Before you code

1. How are stamps ordered for time-based-key-value?
2. What does get return when missing?
3. Can timestamps repeat for time-based-key-value?


```python
import random


class TimeMap:
    def __init__(self):
        raise NotImplementedError

    def set(self, key: str, value: str, timestamp: int) -> None:
        raise NotImplementedError

    def get(self, key: str, timestamp: int) -> str:
        raise NotImplementedError


class _BruteTimeMap:
    def __init__(self):
        self.data = {}

    def set(self, key, value, timestamp):
        self.data.setdefault(key, []).append((timestamp, value))

    def get(self, key, timestamp):
        best = ""
        for t, v in self.data.get(key, []):
            if t <= timestamp:
                best = v
            else:
                break
        return best


tm = TimeMap()
tm.set("foo", "bar", 1)
assert tm.get("foo", 1) == "bar"
assert tm.get("foo", 3) == "bar"
tm.set("foo", "bar2", 4)
assert tm.get("foo", 4) == "bar2"
assert tm.get("foo", 5) == "bar2"
assert tm.get("missing", 1) == ""
tm2 = TimeMap()
tm2.set("a", "x", 2)
assert tm2.get("a", 1) == ""
assert tm2.get("a", 2) == "x"
assert tm2.get("a", 3) == "x"
rng = random.Random(65)
for _ in range(30):
    ref, got = _BruteTimeMap(), TimeMap()
    t = 0
    for __ in range(20):
        t += rng.randint(1, 3)
        if rng.random() < 0.6:
            key = rng.choice("ab")
            val = rng.choice("xyz")
            ref.set(key, val, t)
            got.set(key, val, t)
        else:
            key = rng.choice("ab")
            q = rng.randint(0, t + 2)
            assert got.get(key, q) == ref.get(key, q)
print("all checks passed")
```
