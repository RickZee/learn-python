# 65. Time Based Key-Value Store

**Difficulty:** Medium. **Target:** O(log n) get. **Time box:** 30 minutes.

Build a store with `set(key, value, timestamp)` and `get(key, timestamp)`. Timestamps for a key arrive in non-decreasing order. `get` returns the value with the largest timestamp ≤ the query, or `""` if none exists.

Versioned configs and as-of queries need the same binary search.

**Constraints.** Keys and values are short strings. Random op sequences stay small.


## Examples

- set(`foo`,`bar`,1); get(`foo`,1) → `bar`; get(`foo`,3) → `bar`; set(`foo`,`bar2`,4); get(`foo`,4) → `bar2`; get(`foo`,5) → `bar2`.


## Before you code

1. What do you store per key?
2. Why does non-decreasing set order let you binary search?
3. What does get return when every timestamp is too large?


## Hints

<details><summary>Hint 1: the pattern</summary>

Design: per-key list of (timestamp, value); binary search on get.
</details>

<details><summary>Hint 2: the approach</summary>

bisect_right for the query timestamp, then step one left.
</details>

<details><summary>Hint 3: near the code</summary>

```python
i = bisect.bisect_right(items, (timestamp, ___)) - 1
return items[i][1] if i >= 0 else ""
```
</details>


````python
from pylearn.challenges import cross_check_ops


class TimeMap:
    def __init__(self) -> None:
        raise NotImplementedError

    def set(self, key: str, value: str, timestamp: int) -> None:
        raise NotImplementedError

    def get(self, key: str, timestamp: int) -> str:
        raise NotImplementedError


tm = TimeMap()
tm.set("foo", "bar", 1)
assert tm.get("foo", 1) == "bar"
assert tm.get("foo", 3) == "bar"
tm.set("foo", "bar2", 4)
assert tm.get("foo", 4) == "bar2"
assert tm.get("foo", 5) == "bar2"
assert tm.get("foo", 0) == ""
assert tm.get("missing", 1) == ""
u = TimeMap()
u.set("x", "a", 2)
u.set("x", "b", 5)
assert u.get("x", 3) == "a"
cross_check_ops("65-time-based-key-value", TimeMap)
print("all checks passed")
````
