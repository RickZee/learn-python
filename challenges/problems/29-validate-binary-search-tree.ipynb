# 29. Validate Binary Search Tree

**Difficulty:** Medium. **Target:** O(n) time, O(n) stack space. **Time box:** 25 minutes.

Return whether every node's value is strictly greater than everything in its left subtree and strictly less than everything in its right subtree. Comparing only to the parent is not enough.

Sorted indexes and search trees rely on this global ordering.

**Constraints.** The tree has 0 to 20 nodes with distinct integer values.


## Examples

- `[2,1,3]` → `True`.
- `[5,1,4,None,None,3,6]` → `False` (`3` sits under `4` but is less than `5`).
- `[]` → `True`.
- `[1]` → `True`.

````python
class TreeNode:
    def __init__(
        self,
        val: int = 0,
        left: "TreeNode | None" = None,
        right: "TreeNode | None" = None,
    ) -> None:
        self.val = val
        self.left = left
        self.right = right


def tree(values: list[int | None]) -> TreeNode | None:
    if not values or values[0] is None:
        return None
    root = TreeNode(values[0])
    queue: list[TreeNode] = [root]
    index = 1
    while queue and index < len(values):
        node = queue.pop(0)
        if index < len(values) and values[index] is not None:
            node.left = TreeNode(values[index])  # type: ignore[arg-type]
            queue.append(node.left)
        index += 1
        if index < len(values) and values[index] is not None:
            node.right = TreeNode(values[index])  # type: ignore[arg-type]
            queue.append(node.right)
        index += 1
    return root


def flatten(root: TreeNode | None) -> list[int | None]:
    if root is None:
        return []
    out: list[int | None] = []
    queue: list[TreeNode | None] = [root]
    while queue:
        node = queue.pop(0)
        if node is None:
            out.append(None)
            continue
        out.append(node.val)
        queue.append(node.left)
        queue.append(node.right)
    while out and out[-1] is None:
        out.pop()
    return out


def is_valid_bst(root: TreeNode | None) -> bool:
    raise NotImplementedError


from pylearn.challenges import cross_check

assert is_valid_bst(tree([2, 1, 3])) is True
assert is_valid_bst(tree([5, 1, 4, None, None, 3, 6])) is False
assert is_valid_bst(tree([])) is True
assert is_valid_bst(tree([1])) is True
assert is_valid_bst(tree([5, 4, 6, None, None, 3, 7])) is False
assert is_valid_bst(tree([1, 1])) is False
assert is_valid_bst(tree([10, 5, 15, None, None, 6, 20])) is False
assert is_valid_bst(tree([3, 1, 5, 0, 2, 4, 6])) is True
cross_check("29-validate-binary-search-tree", lambda vals: is_valid_bst(tree(vals)))
print("all checks passed")
````
