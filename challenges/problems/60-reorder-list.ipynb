# 60. Reorder List

**Difficulty:** Medium. **Target:** O(n) time. **Time box:** 25 minutes.

Reorder values `L0 → L1 → … → Ln` into `L0 → Ln → L1 → Ln-1 → …` by interleaving the front half with the reversed back half. Return the new sequence (on real nodes you would relink in place).

Playlist and deck shuffles use the same split-reverse-merge idea.

**Constraints.** Length at most 12 in the random checks.


## Examples

- `[1,2,3,4]` → `[1,4,2,3]`.
- `[1,2,3,4,5]` → `[1,5,2,4,3]`.
- `[1]` → `[1]`.
- `[1,2]` → `[1,2]`.
