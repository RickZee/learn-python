```python
import random


def reorder_values(*args, **kwargs):
    raise NotImplementedError


def _brute_reorder(values):
    vals = list(values)
    if len(vals) <= 2:
        return vals
    mid = (len(vals) + 1) // 2
    left, right = vals[:mid], list(reversed(vals[mid:]))
    out = []
    for i, a in enumerate(left):
        out.append(a)
        if i < len(right):
            out.append(right[i])
    return out


assert reorder_values([1, 2, 3, 4]) == [1, 4, 2, 3]
assert reorder_values([1, 2, 3, 4, 5]) == [1, 5, 2, 4, 3]
assert reorder_values([1]) == [1]
assert reorder_values([1, 2]) == [1, 2]
assert reorder_values([1, 2, 3]) == [1, 3, 2]
assert reorder_values([]) == []
assert reorder_values([1, 2, 3, 4, 5, 6]) == [1, 6, 2, 5, 3, 4]
assert reorder_values([0, 0]) == [0, 0]
rng = random.Random(60)
for _ in range(40):
    values = [rng.randint(0, 9) for _ in range(rng.randint(0, 12))]
    assert reorder_values(values) == _brute_reorder(values)
print("all checks passed")
```
