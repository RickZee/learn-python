```python
import random


def find_redundant(*args, **kwargs):
    raise NotImplementedError


def _brute_redundant(edges):
    parent = {}

    def find(x):
        parent.setdefault(x, x)
        while parent[x] != x:
            parent[x] = parent[parent[x]]
            x = parent[x]
        return x

    for a, b in edges:
        ra, rb = find(a), find(b)
        if ra == rb:
            return [a, b]
        parent[rb] = ra
    return []


assert find_redundant([[1, 2], [1, 3], [2, 3]]) == [2, 3]
assert find_redundant([[1, 2], [2, 3], [3, 4], [1, 4], [1, 5]]) == [1, 4]
assert find_redundant([[1, 2], [2, 3], [3, 1]]) == [3, 1]
assert find_redundant([[1, 2], [2, 3], [3, 4], [4, 1], [1, 5]]) == [4, 1]
assert find_redundant([[1, 2], [1, 3], [2, 3]]) == [2, 3]
assert find_redundant([[2, 3], [1, 2], [1, 3]]) == [1, 3]
assert find_redundant([[1, 2], [2, 3], [1, 3]]) == [1, 3]
assert find_redundant([[1, 2], [3, 4], [2, 3], [1, 4]]) == [1, 4]
rng = random.Random(50)
for _ in range(30):
    n = rng.randint(3, 7)
    edges = [[i, i + 1] for i in range(1, n)]
    extra = [rng.randint(1, n), rng.randint(1, n)]
    while abs(extra[0] - extra[1]) <= 1:
        extra = [rng.randint(1, n), rng.randint(1, n)]
    edges.append(extra)
    assert find_redundant(edges) == _brute_redundant(edges)
print("all checks passed")
```
