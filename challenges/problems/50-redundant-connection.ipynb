# 50. Redundant Connection

**Difficulty:** Medium. **Target:** O(E · α(V)). **Time box:** 25 minutes.

An undirected graph with `n` nodes labeled `1..n` started as a tree, then one extra edge was added. Given the edge list in order, return the edge that can be removed so the graph is a tree again. If several work, return the one that appears last in the input.

Union-find detects the first edge that closes a cycle.

**Constraints.** Exactly one redundant edge. `n` stays small in checks.


## Examples

- `[[1,2],[1,3],[2,3]]` → `[2,3]`.
- `[[1,2],[2,3],[3,4],[1,4],[1,5]]` → `[1,4]`.
