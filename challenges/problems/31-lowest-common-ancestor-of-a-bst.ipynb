# 31. Lowest Common Ancestor of a BST

**Difficulty:** Medium. **Target:** O(h) time, O(1) extra space. **Time box:** 25 minutes.

In a binary search tree, find the lowest common ancestor of values `p` and `q` (a node may be an ancestor of itself). Return that ancestor's value. Both values exist in the tree.

Directory trees with ordered keys use the same split-point walk.

**Constraints.** The tree is a BST with 2 to 20 distinct values. `p` and `q` differ.


## Examples

- Tree `[6,2,8,0,4,7,9,None,None,3,5]`, `p = 2`, `q = 8` → `6`.
- Same tree, `p = 2`, `q = 4` → `2`.
- Tree `[2,1,3]`, `p = 1`, `q = 3` → `2`.
- Tree `[2,1]`, `p = 2`, `q = 1` → `2`.

````python
class TreeNode:
    def __init__(
        self,
        val: int = 0,
        left: "TreeNode | None" = None,
        right: "TreeNode | None" = None,
    ) -> None:
        self.val = val
        self.left = left
        self.right = right


def tree(values: list[int | None]) -> TreeNode | None:
    if not values or values[0] is None:
        return None
    root = TreeNode(values[0])
    queue: list[TreeNode] = [root]
    index = 1
    while queue and index < len(values):
        node = queue.pop(0)
        if index < len(values) and values[index] is not None:
            node.left = TreeNode(values[index])  # type: ignore[arg-type]
            queue.append(node.left)
        index += 1
        if index < len(values) and values[index] is not None:
            node.right = TreeNode(values[index])  # type: ignore[arg-type]
            queue.append(node.right)
        index += 1
    return root


def flatten(root: TreeNode | None) -> list[int | None]:
    if root is None:
        return []
    out: list[int | None] = []
    queue: list[TreeNode | None] = [root]
    while queue:
        node = queue.pop(0)
        if node is None:
            out.append(None)
            continue
        out.append(node.val)
        queue.append(node.left)
        queue.append(node.right)
    while out and out[-1] is None:
        out.pop()
    return out


def lowest_common_ancestor(root: TreeNode | None, p: int, q: int) -> int:
    raise NotImplementedError


from pylearn.challenges import cross_check

t = tree([6, 2, 8, 0, 4, 7, 9, None, None, 3, 5])
assert lowest_common_ancestor(t, 2, 8) == 6
assert lowest_common_ancestor(t, 2, 4) == 2
assert lowest_common_ancestor(tree([2, 1, 3]), 1, 3) == 2
assert lowest_common_ancestor(tree([2, 1]), 2, 1) == 2
assert lowest_common_ancestor(tree([5, 3, 8, 1, 4, 7, 9]), 1, 4) == 3
assert lowest_common_ancestor(tree([5, 3, 8, 1, 4, 7, 9]), 7, 9) == 8
assert lowest_common_ancestor(tree([5, 3, 8, 1, 4, 7, 9]), 3, 8) == 5
assert lowest_common_ancestor(tree([5, 3, 8, 1, 4, 7, 9]), 4, 7) == 5
cross_check(
    "31-lowest-common-ancestor-of-a-bst",
    lambda vals, p, q: lowest_common_ancestor(tree(vals), p, q),
)
print("all checks passed")
````
