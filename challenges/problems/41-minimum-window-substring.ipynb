# 41. Minimum Window Substring

**Difficulty:** Hard. **Target:** O(|s| + |t|) time. **Time box:** 40 minutes.

Return the shortest substring of `s` that covers every character of `t` (including duplicates). If several windows share that length, any shortest one is fine; if none exists, return the empty string.

Search UIs highlight the smallest span that contains a query's letters.

**Constraints.** Strings use lowercase letters. Lengths stay small in checks.


## Examples

- `s = "ADOBECODEBANC"`, `t = "ABC"` → `"BANC"`.
- `s = "a"`, `t = "a"` → `"a"`.
- `s = "a"`, `t = "aa"` → `""`.
- `s = "ab"`, `t = "b"` → `"b"`.


## Before you code

1. When does the window first cover `t`?
2. What do you shrink, and when do you stop shrinking?
3. How do duplicate letters in `t` change the need counts?
