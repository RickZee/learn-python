# 59. Serialize and Deserialize Binary Tree

**Difficulty:** Hard. **Target:** O(n) encode and decode. **Time box:** 30 minutes.

Encode a level-order list of values (with `None` holes) into a string, and decode that string back to the same list. Use the token `null` for missing nodes so the shape is unambiguous.

Persisting a tree across a wire needs a reversible encoding.

**Constraints.** Lists have up to 12 slots in the random checks.


## Examples

- `[1,2,3,None,None,4,5]` ↔ `"1,2,3,null,null,4,5"`.
- `[]` ↔ `""`.
- `[1]` ↔ `"1"`.
