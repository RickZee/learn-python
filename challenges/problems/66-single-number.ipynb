# 66. Single Number

**Difficulty:** Easy. **Target:** O(n) time, O(1) space. **Time box:** 10 minutes.

Every value in `nums` appears twice except one value that appears once. Return that singleton. Solve it with constant extra space.

XOR cancels pairs; the leftover bit pattern is the answer.

**Constraints.** Length is odd and at least 1. Values fit in machine ints.


## Examples

- `[2,2,1]` → `1`.
- `[4,1,2,1,2]` → `4`.
- `[1]` → `1`.
