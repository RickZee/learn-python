# 22. Trapping Rain Water

**Difficulty:** Hard. **Target:** O(n) time, O(1) extra space. **Time box:** 40 minutes.

You are given bar heights `height[i]`. After it rains, how many units of water sit in the valleys between bars? Water at index i is limited by the shorter of the tallest bars to its left and right.

Elevation maps and histogram problems share this two-pointer idea.

**Constraints.** `height` has 0 to 20_000 non-negative integers.


## Examples

- `[0,1,0,2,1,0,1,3,2,1,2,1]` → `6`.
- `[4,2,0,3,2,5]` → `9`.
- `[]` → `0`.
- `[2,0,2]` → `2`.
