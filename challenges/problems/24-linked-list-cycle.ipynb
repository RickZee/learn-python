# 24. Linked List Cycle

**Difficulty:** Easy. **Target:** O(n) time, O(1) extra space. **Time box:** 20 minutes.

You are given the head of a linked list that may contain a cycle. Return whether a cycle exists. Aim for constant extra space.

Floyd's tortoise and hare is the standard constant-space detection.

**Constraints.** The list has 0 to 10_000 nodes. A cycle entry index of -1 means no cycle in the fixtures.


## Examples

- `[3,2,0,-4]` with cycle at index 1 → `True`.
- `[1,2]` with cycle at index 0 → `True`.
- `[1]` with no cycle → `False`.
- `[]` → `False`.


## Before you code

1. Why do slow and fast pointers meet only if a cycle exists?
2. What happens if you store every seen node in a set instead?
3. When should the loop stop if there is no cycle?


## Hints

<details><summary>Hint 1: the pattern</summary>

Floyd cycle detection.
</details>

<details><summary>Hint 2: the approach</summary>

Slow steps one node, fast steps two. They meet inside a cycle; if fast hits None, there is no cycle.
</details>

<details><summary>Hint 3: near the code</summary>

```python
while fast is not None and fast.next is not None:
    slow = slow.next
    fast = fast.next.next
    if slow is ___
        return True
```
</details>


````python
class ListNode:
    def __init__(self, val: int = 0, next: "ListNode | None" = None) -> None:
        self.val = val
        self.next = next


def nodes(values: list[int]) -> ListNode | None:
    dummy = ListNode()
    tail = dummy
    for value in values:
        tail.next = ListNode(value)
        tail = tail.next
    return dummy.next


def values(head: ListNode | None) -> list[int]:
    out: list[int] = []
    while head is not None:
        out.append(head.val)
        head = head.next
    return out


def has_cycle(head: ListNode | None) -> bool:
    raise NotImplementedError


def with_cycle(vals: list[int], pos: int) -> ListNode | None:
    head = nodes(vals)
    if not vals:
        return head
    arr: list[ListNode] = []
    cur = head
    while cur is not None:
        arr.append(cur)
        cur = cur.next
    if 0 <= pos < len(arr):
        arr[-1].next = arr[pos]
    return head


from pylearn.challenges import cross_check

assert has_cycle(with_cycle([3, 2, 0, -4], 1)) is True
assert has_cycle(with_cycle([1, 2], 0)) is True
assert has_cycle(with_cycle([1], -1)) is False
assert has_cycle(with_cycle([], -1)) is False
assert has_cycle(with_cycle([1, 2, 3], -1)) is False
assert has_cycle(with_cycle([1, 2, 3], 2)) is True
assert has_cycle(with_cycle([1], 0)) is True
assert has_cycle(with_cycle([1, 2, 3, 4], 1)) is True
cross_check("24-linked-list-cycle", lambda vals, pos: has_cycle(with_cycle(vals, pos)))
print("all checks passed")
````
