# 18. House Robber

**Difficulty:** Medium. **Target:** O(n) time, O(1) extra space. **Time box:** 20 minutes.

You are a robber walking a street of houses with values `nums[i]`. You cannot rob two adjacent houses. Return the maximum total you can rob.

Choosing non-adjacent slots is a classic 1D DP decision.

**Constraints.** `nums` has 0 to 100 non-negative integers.


## Examples

- `[1,2,3,1]` → `4` (1+3).
- `[2,7,9,3,1]` → `12`.
- `[]` → `0`.
- `[5]` → `5`.
