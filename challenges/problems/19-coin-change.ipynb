# 19. Coin Change

**Difficulty:** Medium. **Target:** O(amount · len(coins)) time. **Time box:** 30 minutes.

You are given coin denominations `coins` and a total `amount`. Return the fewest coins that sum to `amount`, or `-1` if it is impossible. You may use each denomination any number of times.

Making change with a limited set of coins is unbounded knapsack.

**Constraints.** Up to 12 denominations. `amount` is between 0 and 10_000.


## Examples

- `[1,2,5]`, `11` → `3` (5+5+1).
- `[2]`, `3` → `-1`.
- `[1]`, `0` → `0`.
- `[1,2,5]`, `100` → `20`.
