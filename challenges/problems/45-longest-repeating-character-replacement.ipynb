```python
import random


def character_replacement(*args, **kwargs):
    raise NotImplementedError


def _brute_character_replacement(s, k):
    best = 0
    for left in range(len(s)):
        counts = {}
        for right in range(left, len(s)):
            counts[s[right]] = counts.get(s[right], 0) + 1
            need = (right - left + 1) - max(counts.values())
            if need <= k:
                best = max(best, right - left + 1)
    return best


assert character_replacement("ABAB", 2) == 4
assert character_replacement("AABABBA", 1) == 4
assert character_replacement("AAAA", 0) == 4
assert character_replacement("ABCD", 0) == 1
assert character_replacement("A", 1) == 1
assert character_replacement("ABBB", 2) == 4
assert character_replacement("BAAA", 0) == 3
assert character_replacement("ABCDE", 1) == 2
rng = random.Random(45)
for _ in range(40):
    s = "".join(rng.choice("ABCD") for _ in range(rng.randint(0, 10)))
    k = rng.randint(0, 4)
    assert character_replacement(s, k) == _brute_character_replacement(s, k)
print("all checks passed")
```
