# 16. LRU Cache

**Difficulty:** Medium. **Target:** O(1) get and put. **Time box:** 40 minutes.

Design a cache with capacity `capacity`. `get(key)` returns the value or `-1`. `put(key, value)` inserts or updates. Both count as a use. When full, `put` evicts the least recently used key before inserting.

Caches in browsers and databases use the same eviction rule.

**Constraints.** Capacity is at least 1. Keys and values are integers.


## Examples

- capacity 2: put(1,1), put(2,2), get(1)→1, put(3,3) evicts 2, get(2)→-1.
- Updating a key counts as a use.
- get on a missing key returns -1.
- capacity 1: a second put evicts the first key.


## Before you code

1. What structure gives O(1) lookup by key and O(1) move-to-recent?
2. When you `get` a hit, where does that key move?
3. What do you evict when `put` exceeds capacity?


## Hints

<details><summary>Hint 1: the pattern</summary>

</details>

<details><summary>Hint 2: the approach</summary>

Keep keys in use order. On get/put of an existing key, move it to most-recent. On overflow, drop the oldest.
</details>
