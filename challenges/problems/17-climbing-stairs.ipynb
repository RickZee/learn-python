**Difficulty:** Easy. **Target:** O(n) time, O(1) extra space. **Time box:** 15 minutes.

This is the Fibonacci recurrence in disguise. Count sequences of 1-steps and 2-steps; the order of steps matters for each way.

**Constraints.** `n` is between 1 and 45.


## Examples

- `3` → `3`.
