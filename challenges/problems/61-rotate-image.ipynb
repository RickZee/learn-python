# 61. Rotate Image

**Difficulty:** Medium. **Target:** O(n²) time, O(1) space. **Time box:** 20 minutes.

Rotate an `n` by `n` matrix 90° clockwise. Mutate it in place and return the same matrix object.

Image editors and game boards rotate grids this way.

 Rings of four cells cycle together; a fresh matrix copy also works if space is free.

**Constraints.** The matrix is square. `n` is at most 6 in the random checks.


## Examples

- `[[1,2,3],[4,5,6],[7,8,9]]` → `[[7,4,1],[8,5,2],[9,6,3]]`.
- `[[1,2],[3,4]]` → `[[3,1],[4,2]]`.
- `[[1]]` → `[[1]]`.
