```python
import random


def min_eating_speed(*args, **kwargs):
    raise NotImplementedError


def _brute_min_speed(piles, h):
    def hours(speed):
        return sum((p + speed - 1) // speed for p in piles)

    for speed in range(1, max(piles) + 1):
        if hours(speed) <= h:
            return speed
    return max(piles)


assert min_eating_speed([3, 6, 7, 11], 8) == 4
assert min_eating_speed([30, 11, 23, 4, 20], 5) == 30
assert min_eating_speed([30, 11, 23, 4, 20], 6) == 23
assert min_eating_speed([1], 1) == 1
assert min_eating_speed([1, 1, 1, 1], 4) == 1
assert min_eating_speed([100], 2) == 50
assert min_eating_speed([5, 5, 5], 3) == 5
assert min_eating_speed([2, 2], 2) == 2
rng = random.Random(46)
for _ in range(30):
    piles = [rng.randint(1, 20) for _ in range(rng.randint(1, 6))]
    h = rng.randint(len(piles), len(piles) + 30)
    assert min_eating_speed(piles, h) == _brute_min_speed(piles, h)
print("all checks passed")
```
