# 46. Koko Eating Bananas

**Difficulty:** Medium. **Target:** O(n log M) time. **Time box:** 25 minutes.

Koko has piles of bananas and `h` hours. Each hour she chooses one pile and eats at integer speed `k` bananas (a smaller pile still costs a full hour). Return the minimum speed that finishes every pile within `h` hours.

Binary search on the answer fits any monotone feasibility check.

**Constraints.** Piles are positive. `h` is at least `len(piles)`.


## Examples

- `[3,6,7,11]`, `h = 8` → `4`.
- `[30,11,23,4,20]`, `h = 5` → `30`.
- `[30,11,23,4,20]`, `h = 6` → `23`.
- `[1]`, `h = 1` → `1`.
