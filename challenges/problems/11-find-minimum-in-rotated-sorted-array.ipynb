# 11. Find Minimum in Rotated Sorted Array

**Difficulty:** Medium. **Target:** O(log n) time. **Time box:** 20.

You are given a list of distinct integers that was sorted ascending and then rotated. Return the minimum value. Aim for logarithmic time.

Finding the rotation point is the same shape as searching a rotated array.

**Constraints.** `nums` has 1 to 5000 distinct integers.


## Examples

- `[3,4,5,1,2]` → `1`.
- `[4,5,6,7,0,1,2]` → `0`.
- `[11,13,15,17]` → `11` (no rotation).
- `[2,1]` → `1`.
