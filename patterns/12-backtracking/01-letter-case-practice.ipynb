```python
def subsets(nums: list[int]) -> list[list[int]]:
    raise NotImplementedError


def listed(nums: list[int]) -> list[tuple[int, ...]]:
    return sorted(tuple(row) for row in subsets(nums))


check(listed, [1, 2], expect=[(), (1,), (1, 2), (2,)])
check(listed, [], expect=[()])
```
