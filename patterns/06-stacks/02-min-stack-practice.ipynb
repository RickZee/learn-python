## Stretch

Same idea, one change.


```python
class MinStack:
    def push(self, value: int) -> None:
        raise NotImplementedError

    def pop(self) -> None:
        raise NotImplementedError

    def top(self) -> int:
        raise NotImplementedError

    def get_min(self) -> int:
        raise NotImplementedError


def minimum_after() -> int:
    stack = MinStack()
    stack.push(3)
    stack.push(1)
    stack.pop()
    return stack.get_min()


check(minimum_after, expect=3)
```
