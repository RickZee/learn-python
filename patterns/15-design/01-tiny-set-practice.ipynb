## Core

`get` returns the value or `-1`. `get` and `put` both count as use. When `put` would pass capacity, drop the least recently used key. Capacity is at least 1.

<details><summary>Hint: the idea</summary>

Keep keys in use order. The front is the oldest.
</details>

<details><summary>Hint: the approach</summary>

`OrderedDict.move_to_end` marks a use. `popitem(last=False)` drops the oldest.
</details>

<details><summary>Hint: near the code</summary>

On `put`, move the key if it is already there, store the value, then drop if `len` passed capacity.
</details>

## Stretch

`survived` puts two keys, uses the first, then puts a third. Return `get` of the second key.


```python
def survived() -> bool:
    box = TinySet()  # noqa: F821
    box.add(1)
    box.remove(1)
    return box.contains(1)


def refreshed() -> bool:
    box = TinySet()  # noqa: F821
    box.add(1)
    return box.contains(1)


check(survived, expect=False)
check(refreshed, expect=True)
```
